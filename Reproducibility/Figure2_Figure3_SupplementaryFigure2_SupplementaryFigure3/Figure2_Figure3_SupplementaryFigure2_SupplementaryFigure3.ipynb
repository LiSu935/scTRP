```python
# plot_comparison_chiffelle

#!/usr/bin/env python3
"""
plot_comparison_chiffelle.py

Comparison figures across ALL 7 datasets (6 held-out + Chiffelle):
  heatmap + strip+box + summary bar.
Replicates comparison_figure_patient_wise.ipynb with the best Chiffelle config.

Best config for Chiffelle:
  clonotype-level : pool_type=max + ensemble=rank_majority
  cell-level      : pool_type=raw + ensemble=rank_majority
For 6 held-out studies: scTRP_simclr single-ck result from the same pool_type.

Methods:
  6 benchmark methods × 7 datasets
  scTRP (+ TCR clonotype) × 7 datasets  ← primary comparison; "scTRP" in figures

Method order top→bottom in both heatmap and strip+box:
  Caushi(Seurat) → Hanada → Lowery(AUCell) → Meng(UCell) → Oliveira(UCell) → Zheng → scTRP

Reads from : {in_root}/all_patient_wise.csv
Writes to  : {in_root}/figures/comparison_best/{cell,clonotype}/

Usage:
    python plot_comparison_chiffelle.py [--in_root DIR] [--levels cell,clonotype]
"""

import argparse
import warnings
from pathlib import Path

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.stats import wilcoxon as _wilcoxon_fn
from scipy.stats import ttest_rel
from scipy.stats import ttest_ind

warnings.filterwarnings("ignore")

# ── Publication rcParams ──────────────────────────────────────────────────────
matplotlib.rcParams.update({
    "font.family":      "sans-serif",
    "font.sans-serif":  ["Arial", "Helvetica", "DejaVu Sans"],
    "font.size":        24,
    "axes.labelsize":   20,
    "axes.titlesize":   20,
    "xtick.labelsize":  18,
    "ytick.labelsize":  18,
    "legend.fontsize":  18,
    "figure.dpi":       150,
    "savefig.dpi":      300,
    "savefig.bbox":     "tight",
    "pdf.fonttype":     42,
    "ps.fonttype":      42,
    "axes.spines.top":  False,
    "axes.spines.right": False,
})

# ── Constants ─────────────────────────────────────────────────────────────────
BASE = "/cluster/pixstor/xudong-lab/suli/Alg_development/scRNA_TCRBCR_surfaceProtein"
DEFAULT_IN_ROOT = (
    f"{BASE}/results/classifier/case_study/"
    "ICB_chiffelle_5case_inference_6AllasTrain/unified_cell_predictions"
)

POOL_CLONO  = "max"
POOL_CELL   = "raw"
ENSEMBLE    = "rankMin_study" # "rankMin_patient", "zscoreMean_studyPerPatient", "rank_majority", "rankProduct_study"， “rankMin_study”

SIX_STUDIES = ["caushi", "hanada", "lowery", "meng", "oliveira", "zheng"]
ENSEMBLE_STUDIES = ["chiffelle", "moffitt"]
STUDY_ORDER = SIX_STUDIES + ENSEMBLE_STUDIES
STUDY_LABELS = {
    "caushi":    "Caushi\n(NSCLC)",
    "hanada":    "Hanada\n(NSCLC)",
    "lowery":    "Lowery\n(Mel.)",
    "meng":      "Meng\n(PDAC)",
    "oliveira":  "Oliveira\n(Mel.)",
    "zheng":     "Zheng\n(GI)",
    "chiffelle": "Chiffelle\n(Mel.)",
    "moffitt": "Moffitt\n(NSCLC)",
}

# Top→bottom order in heatmap (and matched in strip+box via reversed y_pos)
BENCH_KEYS = [
    "benchmark_Seurat_Caushi", "benchmark_Hanada", "benchmark_AUCell_Lowery",
    "benchmark_UCell_Meng", "benchmark_UCell_Oliveira", "benchmark_Zheng",
    "benchmark_Manascore", "benchmark_Trace",


]
SCTRP_KEY    = "scTRP"          # unified display name
METHOD_ORDER = [SCTRP_KEY] + BENCH_KEYS 

METHOD_LABELS = {
    "benchmark_Seurat_Caushi":  "Caushi et al.\n(Seurat)",
    "benchmark_Hanada":         "Hanada et al.",
    "benchmark_AUCell_Lowery":  "NeoTCR8\n(AUCell)",
    "benchmark_UCell_Meng":     "TR30\n(UCell)",
    "benchmark_UCell_Oliveira": "Oliveira et al.\n(Seurat)",
    "benchmark_Zheng":          "Zheng et al.",
    "benchmark_Manascore": "MANAscore",
    "benchmark_Trace": "TRACE",
    SCTRP_KEY:                  "scTRP",
}

METRICS = [
    ("ROC_AUC", "ROC-AUC",  0.4, 1.0),
    ("MCC",     "MCC",      -0.1, 1.0),
    ("F1",      "F1-score", 0.3, 1.0),
    ("Gmean",   "G-mean",   0.3, 1.0),
]

# Color palette: muted blues for benchmarks, red for scTRP
_bench_pal = sns.color_palette("Blues_d", n_colors=len(BENCH_KEYS) + 2)[2:]
METHOD_PALETTE = {k: c for k, c in zip(BENCH_KEYS, _bench_pal)}
METHOD_PALETTE[SCTRP_KEY] = "#d62728"

ANNOT_PAD = 0.30   # extra x space for Wilcoxon annotation column

# ── Data loading ──────────────────────────────────────────────────────────────

def load_all_patient_wise(in_root):
    csv = Path(in_root) / "all_patient_wise.csv"
    if not csv.exists():
        raise FileNotFoundError(
            f"{csv} not found — run compute_patient_metrics_all.py first"
        )
    df = pd.read_csv(csv)
    print(f"Loaded {len(df):,} rows  |  "
          f"pool_types: {sorted(df['pool_type'].unique())}")
    return df


def build_level_df(df_full, level):
    """
    Select rows for one level ('cell' or 'clonotype') and unify method names.

    Chiffelle uses scTRP_simclr_rank_majority; 6 studies use scTRP_simclr.
    Both remapped to SCTRP_KEY for uniform plotting.
    Benchmarks always come from raw pool (pooling is identity for them).
    """
    pool = POOL_CELL if level == "cell" else POOL_CLONO

    # Benchmark methods for all 7 studies (raw pool)
    bench = df_full[
        (df_full["pool_type"] == pool) &
        (df_full["method"].str.startswith("benchmark_"))
    ].copy()

    # scTRP — 6 held-out studies (single-ck, no mode suffix)
    s_six = df_full[
        (df_full["pool_type"] == pool) &
        (df_full["method"] == "scTRP_simclr") &
        (df_full["study"].isin(SIX_STUDIES))
    ].copy()
    s_six["method"] = SCTRP_KEY

    # scTRP — Chiffelle (6-ck ensemble with mode suffix)
    s_chif = df_full[
        (df_full["pool_type"] == pool) &
        (df_full["method"] == f"scTRP_simclr_{ENSEMBLE}") &
        (df_full["study"].isin(ENSEMBLE_STUDIES))
    ].copy()
    s_chif["method"] = SCTRP_KEY

    combined = pd.concat([bench, s_six, s_chif], ignore_index=True)
    combined["level"] = level

    n = len(combined)
    meth = sorted(combined["method"].unique())
    stud = sorted(combined["study"].unique())
    print(f"  level={level}  rows={n:,}  methods={meth}  studies={stud}")
    return combined


# ── Wilcoxon helper ───────────────────────────────────────────────────────────

def _wilcoxon_stars(a_df, b_df, metric):
    """Paired one-sided Wilcoxon (a > b); DataFrames indexed by sample."""
    common = a_df.index.intersection(b_df.index)
    if len(common) < 3:
        return "n/a", np.nan
    a = a_df.loc[common, metric].values.astype(float)
    b = b_df.loc[common, metric].values.astype(float)
    if np.all(a == b):
        return "ns", 1.0
    try:
        _, p = ttest_rel(a, b, alternative="greater")
    except Exception:
        return "n/a", np.nan
    if p < 0.001: return "***", p
    if p < 0.01:  return "**",  p
    if p < 0.05:  return "*",   p
    return "ns", p


# ── Plot: Heatmap ─────────────────────────────────────────────────────────────

def plot_heatmap(df_all, fig_dir, level):
    """
    Row order (top→bottom): BENCH_KEYS then scTRP.
    Dashed lines bracket the scTRP row.
    """
    fig_dir.mkdir(parents=True, exist_ok=True)
    mord = [m for m in METHOD_ORDER if m in df_all["method"].unique()]
    sord = [s for s in STUDY_ORDER if s in df_all["study"].unique()]
    sctrp_idx = mord.index(SCTRP_KEY) if SCTRP_KEY in mord else None

    for metric, title, vmin, vmax in METRICS:
        if metric not in df_all.columns or df_all[metric].isna().all():
            continue

        pivot = (df_all.groupby(["method", "study"])[metric].median()
                 .unstack("study")
                 .reindex(index=mord, columns=sord))

        nrow, ncol = len(mord), len(sord)
        fig, ax = plt.subplots(figsize=(ncol * 1.9, nrow * 1.05),
                               constrained_layout=True)
        im = ax.imshow(pivot.values.astype(float), cmap="RdYlGn_r",
                       vmin=vmin, vmax=vmax, aspect="auto")

        for i in range(nrow):
            for j in range(ncol):
                val = pivot.values[i, j]
                if not np.isnan(val):
                    fw = "bold" if mord[i] == SCTRP_KEY else "normal"
                    ax.text(j, i, f"{val:.2f}", ha="center", va="center",
                            fontsize=20, color="black", fontweight=fw)
                else:
                    ax.text(j, i, "–", ha="center", va="center",
                            fontsize=11, color="#888888")

        if sctrp_idx is not None:
            ax.axhline(sctrp_idx - 0.5, color="#333", lw=1.5, ls="--")
            ax.axhline(sctrp_idx + 0.5, color="#333", lw=1.5, ls="--")

        ax.set_xticks(range(ncol))
        ax.xaxis.tick_top()
        ax.xaxis.set_label_position("top")
        ax.set_xticklabels([STUDY_LABELS.get(s, s) for s in sord],
                            rotation=0, ha="center")
        ax.set_yticks(range(nrow))
        ax.set_yticklabels([METHOD_LABELS.get(m, m) for m in mord])
        ax.set_xlabel("Study", labelpad=8)
        ax.set_ylabel("Method", labelpad=8)
        ax.set_title(f"Median patient-wise {title}  [{level}]",
                     pad=10, fontweight="bold")

        cbar = fig.colorbar(im, ax=ax, fraction=0.03, pad=0.03, shrink=0.85)
        cbar.set_label(metric, fontsize=20)
        cbar.ax.tick_params(labelsize=18)

        fname = f"heatmap_{metric}_{level}"
        fig.savefig(fig_dir / f"{fname}.pdf")
        fig.savefig(fig_dir / f"{fname}.png", dpi=300)
        plt.show(fig)
        plt.close(fig)
        print(f"    saved {fname}")


# ── Plot: Strip + Box ─────────────────────────────────────────────────────────

def plot_stripbox(df_all, fig_dir, level):
    """
    One panel per study, grouped into two blocks: 6 leave-one-out test
    studies (left) and independent test studies (right), each block
    underlined with its own title bar.
    Horizontal strip+box per method, y-axis top→bottom matches heatmap:
      Caushi → Hanada → Lowery → Meng → Oliveira → Zheng → scTRP
    Paired t-test (scTRP > each benchmark) annotated at right edge, aligned
    per study using that study's own data range.
    """
    fig_dir.mkdir(parents=True, exist_ok=True)
    mord   = [m for m in METHOD_ORDER if m in df_all["method"].unique()]
    sord   = [s for s in STUDY_ORDER if s in df_all["study"].unique()]
    n_meth = len(mord)

    study_6     = [s for s in SIX_STUDIES if s in sord]
    study_indep = [s for s in sord if s not in study_6]
    n_left, n_right = len(study_6), len(study_indep)

    # Reversed y positions: METHOD_ORDER[0] (Seurat_Caushi) → highest y (top)
    y_pos = {m: n_meth - 1 - i for i, m in enumerate(mord)}

    for metric, ylabel, xmin, xmax in METRICS:
        if metric not in df_all.columns or df_all[metric].isna().all():
            continue

        fig = plt.figure(figsize=((n_left + n_right) * 2.5 + 0.5,
                                   n_meth * 0.75 + 2.0),
                          constrained_layout=True)
        width_ratios = [1] * n_left + [0.15] + [1] * n_right
        gs = fig.add_gridspec(1, n_left + 1 + n_right, width_ratios=width_ratios)

        axes, ax0 = [], None
        for i, study in enumerate(study_6):
            ax = fig.add_subplot(gs[0, i], sharey=ax0)
            ax0 = ax0 or ax
            axes.append((ax, study))

        # spacer column between the two blocks (kept invisible)
        div_ax = fig.add_subplot(gs[0, n_left])
        div_ax.axis("off")

        for j, study in enumerate(study_indep):
            ax = fig.add_subplot(gs[0, n_left + 1 + j], sharey=ax0)
            axes.append((ax, study))

        for ax, study in axes:
            sub = df_all[df_all["study"] == study]
            sctrp_df = (sub[sub["method"] == SCTRP_KEY][["sample", metric]]
                        .dropna().set_index("sample"))

            # ── Pass 1: compute this study's final data range across ALL methods ──
            data_min, data_max = np.inf, -np.inf
            for method in mord:
                yvals = sub[sub["method"] == method][metric].dropna().values
                if len(yvals) == 0:
                    continue
                data_min = min(data_min, float(yvals.min()))
                data_max = max(data_max, float(yvals.max()))
            if not np.isfinite(data_min):
                data_min, data_max = xmin, xmax
            pad_left = max((data_max - data_min) * 0.10, 0.015)
            star_x = data_max + ANNOT_PAD * 0.95   # fixed for the whole column

            # ── Pass 2: draw bars, scatter, and stars using the fixed range ────────
            rng = np.random.RandomState(42)
            for method in mord:
                msub  = sub[sub["method"] == method]
                yvals = msub[metric].dropna().values
                if len(yvals) == 0:
                    continue
                yp    = y_pos[method]
                color = METHOD_PALETTE.get(method, "#aaa")

                if len(yvals) >= 2:
                    q1, q2, q3 = np.percentile(yvals, [25, 50, 75])
                    iqr = q3 - q1
                    lo  = max(yvals.min(), q1 - 1.5 * iqr)
                    hi  = min(yvals.max(), q3 + 1.5 * iqr)
                    ax.barh(yp, q3 - q1, left=q1, height=0.70,
                            color=color, alpha=0.35, zorder=2)
                    ax.vlines(q2, yp - 0.35, yp + 0.35, color=color, lw=3.5, zorder=3)
                    ax.hlines(yp, lo, hi, color=color, lw=1.8, alpha=0.7, zorder=2)

                is_sctrp = method == SCTRP_KEY
                jit = rng.uniform(-0.18, 0.18, size=len(yvals))
                ax.scatter(yvals, yp + jit,
                           color=color,
                           s=(6.0 if is_sctrp else 4.5) ** 2.5,
                           alpha=0.90 if is_sctrp else 0.65,
                           zorder=4,
                           edgecolors="white" if is_sctrp else "none",
                           linewidths=0.5)

                # Wilcoxon: scTRP > this benchmark, star aligned via star_x
                if method != SCTRP_KEY and len(sctrp_df) >= 3:
                    bench_df = msub[["sample", metric]].dropna().set_index("sample")
                    star, _  = _wilcoxon_stars(sctrp_df, bench_df, metric)
                    if star not in ["ns", "n/a"]:
                        ax.text(star_x, yp, star,
                                ha="right", va="center", fontsize=22, color="#333")

            # Subtle highlight behind scTRP row
            if SCTRP_KEY in y_pos:
                sy = y_pos[SCTRP_KEY]
                ax.axhspan(sy - 0.45, sy + 0.45, color="#d62728", alpha=0.07, zorder=0)

            #ax.axvline(0.5, color="gray", lw=0.8, ls="--", zorder=1)
            ax.set_xlim(data_min - pad_left, data_max + ANNOT_PAD)
            ax.set_ylim(-0.5, n_meth - 0.5)
            ax.set_title(STUDY_LABELS.get(study, study), pad=5, fontsize=22)
            ax.tick_params(axis="x", labelsize=20)

        # Y-tick labels: bottom→top = METHOD_ORDER reversed
        axes[0][0].set_yticks(range(n_meth))
        axes[0][0].set_yticklabels(
            [METHOD_LABELS.get(m, m) for m in reversed(mord)], fontsize=22
        )
        axes[0][0].set_ylabel("Method", labelpad=8, fontweight="bold", fontsize=24)
        for ax, _ in axes[1:]:
            plt.setp(ax.get_yticklabels(), visible=False)

        fig.suptitle(f"Patient-wise {ylabel} — scTRP vs. benchmark  [{level}]",
                     fontsize=24, fontweight="bold", y=1.1)

        # ── Group titles with horizontal underline bars (no vertical divider) ──
        fig.canvas.draw()
        left_pos  = [ax.get_position() for ax, _ in axes[:n_left]]
        right_pos = [ax.get_position() for ax, _ in axes[n_left:]]
        top_y = max(p.y1 for p in left_pos + right_pos)

        left_x0, left_x1   = left_pos[0].x0, left_pos[-1].x1
        right_x0, right_x1 = right_pos[0].x0, right_pos[-1].x1

        title_y = top_y + 0.105
        line_y  = top_y + 0.09

        fig.text((left_x0 + left_x1) / 2, title_y,
                 "Six leave-one-out test studies", ha="center",
                 fontsize=22, fontweight="bold")
        fig.add_artist(plt.Line2D([left_x0, left_x1], [line_y, line_y],
                                  color="#999999", lw=2.8,
                                  transform=fig.transFigure, clip_on=False))

        fig.text((right_x0 + right_x1) / 2, title_y,
                 "Independent test studies", ha="center",
                 fontsize=22, fontweight="bold")
        fig.add_artist(plt.Line2D([right_x0, right_x1], [line_y, line_y],
                                  color="#999999", lw=2.8,
                                  transform=fig.transFigure, clip_on=False))

        fname = f"stripbox_{metric}_{level}"
        fig.savefig(fig_dir / f"{fname}.pdf", bbox_inches="tight")
        fig.savefig(fig_dir / f"{fname}.png", dpi=300, bbox_inches="tight")
        plt.show(fig)
        plt.close(fig)
        print(f"    saved {fname}")


# ── Plot: scTRP vs best benchmark summary bar ─────────────────────────────────

def plot_bar_vs_best_benchmark(df_all, fig_dir, level):
    """Grouped bar: best-benchmark vs scTRP per study (ROC_AUC)."""
    fig_dir.mkdir(parents=True, exist_ok=True)
    bench_present = [m for m in BENCH_KEYS if m in df_all["method"].unique()]
    sord = [s for s in STUDY_ORDER if s in df_all["study"].unique()]

    metric, title, vmin, vmax = METRICS[0]   # ROC_AUC
    if metric not in df_all.columns:
        return

    rows = []
    for study in sord:
        sub = df_all[df_all["study"] == study]
        sctrp_m = sub[sub["method"] == SCTRP_KEY][metric].mean()
        best_b  = (sub[sub["method"].isin(bench_present)]
                   .groupby("method")[metric].mean().max()
                   if bench_present else np.nan)
        rows.append({"study": study, "scTRP": sctrp_m, "best_bench": best_b})
    df_s = pd.DataFrame(rows).set_index("study").reindex(sord)

    fig, ax = plt.subplots(figsize=(len(sord) * 1.3, 3.5))
    x = np.arange(len(sord))
    w = 0.35
    b1 = ax.bar(x - w / 2, df_s["best_bench"], width=w, color="#4878CF", alpha=0.8,
                label="Best benchmark", zorder=3)
    b2 = ax.bar(x + w / 2, df_s["scTRP"],      width=w, color="#d62728", alpha=0.9,
                label="scTRP (ours)", zorder=3)

    for bar in (*b1, *b2):
        h = bar.get_height()
        if not np.isnan(h):
            ax.text(bar.get_x() + bar.get_width() / 2, h + 0.005, f"{h:.2f}",
                    ha="center", va="bottom", fontsize=9)

    ax.axhline(0.5, color="gray", lw=0.8, ls="--")
    ax.set_xticks(x)
    ax.set_xticklabels([STUDY_LABELS.get(s, s) for s in sord], ha="center")
    ax.set_ylabel(f"Mean patient-wise {title}")
    ax.set_ylim(max(vmin - 0.1, 0.0), min(vmax + 0.15, 1.08))
    ax.set_title(f"scTRP vs best benchmark — {title}  [{level}]",
                 fontweight="bold", pad=8)
    ax.legend(frameon=False, loc="lower right", fontsize=14)
    ax.yaxis.grid(True, lw=0.5, color="lightgray", zorder=0)

    plt.tight_layout()
    fname = f"bar_scTRP_vs_best_benchmark_{level}"
    fig.savefig(fig_dir / f"{fname}.pdf")
    fig.savefig(fig_dir / f"{fname}.png", dpi=300)
    plt.show(fig)
    plt.close(fig)
    print(f"    saved {fname}")


# ── Wilcoxon significance table ───────────────────────────────────────────────

def save_wilcoxon_table(df_all, fig_dir, level):
    bench_present = [m for m in BENCH_KEYS if m in df_all["method"].unique()]
    sord  = [s for s in STUDY_ORDER if s in df_all["study"].unique()]
    rows  = []

    for metric, *_ in METRICS:
        if metric not in df_all.columns:
            continue
        for method in bench_present:
            for study in sord:
                sub   = df_all[df_all["study"] == study]
                sctrp = (sub[sub["method"] == SCTRP_KEY][["sample", metric]]
                         .dropna().set_index("sample"))
                bench = (sub[sub["method"] == method][["sample", metric]]
                         .dropna().set_index("sample"))
                star, p = _wilcoxon_stars(sctrp, bench, metric)
                rows.append({"Metric": metric, "Method": method, "Study": study,
                             "n_pairs": len(sctrp.index.intersection(bench.index)),
                             "p_value": p, "sig": star})

    pd.DataFrame(rows).to_csv(
        fig_dir / f"sig_scTRP_vs_benchmark_{level}.csv", index=False
    )
    print(f"    saved sig_scTRP_vs_benchmark_{level}.csv")


```

```python
#!/usr/bin/env python3
"""
plot_ablation_chiffelle.py

Ablation figures: scTRP (+ TCR clonotype) vs scTRP (transcription only).
Replicates ablation_scTRP_clonotype_vs_transcription.ipynb +
topn_precision_chiffelle.ipynb with the best Chiffelle config.

Best config for Chiffelle:
  clonotype-level : pool_type=max + ensemble=rank_majority
  cell-level      : pool_type=raw + ensemble=rank_majority
For 6 held-out studies: scTRP_simclr / scTRP_only single-ck from same pool_type.

Figures produced per level (cell / clonotype):
  Fig 1 — Heatmap: ablation trio × 7 datasets
  Fig 2 — Grouped barplot: mean metric per study (6 studies | Chiffelle)
  Fig 3 — Δ bar chart: (+ TCR clonotype) − (transcription only) per study
  Fig 4 — Per-patient strip+box on Chiffelle (T-test)
  Fig 5 — Precision@k bar chart on Chiffelle

Reads from : {in_root}/all_patient_wise.csv
             {in_root}/raw_cell_scores/ and pooled_scores/max/
Writes to  : {in_root}/figures/ablation_best/{cell,clonotype}/

Usage:
    python plot_ablation_chiffelle.py [--in_root DIR] [--levels cell,clonotype]
"""

import argparse
import warnings
from collections import defaultdict
from pathlib import Path

import matplotlib
import matplotlib.patches as mpatches
import matplotlib.ticker as mticker
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.stats import wilcoxon as _wilcoxon_fn
from sklearn.metrics import (
    balanced_accuracy_score, confusion_matrix, f1_score,
    matthews_corrcoef, roc_auc_score,
)

try:
    import jenkspy
    HAS_JENKSPY = True
except ImportError:
    HAS_JENKSPY = False

warnings.filterwarnings("ignore")

# ── Publication rcParams ──────────────────────────────────────────────────────
matplotlib.rcParams.update({
    "font.family":        "sans-serif",
    "font.sans-serif":    ["Arial", "Helvetica", "DejaVu Sans"],
    "font.size":          14,
    "axes.labelsize":     18,
    "axes.titlesize":     18,
    "axes.labelweight":   "bold",
    "axes.titleweight":   "bold",
    "xtick.labelsize":    15,
    "ytick.labelsize":    15,
    "legend.fontsize":    14,
    "figure.dpi":         150,
    "savefig.dpi":        1200,
    "savefig.bbox":       "tight",
    "pdf.fonttype":       42,
    "ps.fonttype":        42,
    "axes.linewidth":     1.2,
    "axes.spines.top":    False,
    "axes.spines.right":  False,
    "xtick.major.width":  1.2,
    "ytick.major.width":  1.2,
    "xtick.major.size":   5,
    "ytick.major.size":   5,
    "lines.linewidth":    2.0,
})

# ── Constants ─────────────────────────────────────────────────────────────────
BASE = "/cluster/pixstor/xudong-lab/suli/Alg_development/scRNA_TCRBCR_surfaceProtein"
DEFAULT_IN_ROOT = (
    f"{BASE}/results/classifier/case_study/"
    "ICB_chiffelle_5case_inference_6AllasTrain/unified_cell_predictions"
)

POOL_CLONO  = "max"
POOL_CELL   = "raw"
ENSEMBLE    = "rankMin_study" #"zscoreMean_studyPerPatient" #"rankMin_patient" # "rankProduct_study"， "rankMin_study"

SIX_STUDIES = ["caushi", "hanada", "lowery", "meng", "oliveira", "zheng"]
ENSEMBLE_STUDIES = ["chiffelle", "moffitt"]
STUDY_ORDER = SIX_STUDIES + ENSEMBLE_STUDIES
STUDY_LABELS = {
    "caushi":    "Caushi\n(NSCLC)",
    "hanada":    "Hanada\n(NSCLC)",
    "lowery":    "Lowery\n(Mel.)",
    "meng":      "Meng\n(PDAC)",
    "oliveira":  "Oliveira\n(Mel.)",
    "zheng":     "Zheng\n(GI)",
    "chiffelle": "Chiffelle\n(Mel.)",
    "moffitt": "moffitt\n(NSCLC)",
}

# Unified method keys after remapping
KEY_CLONO = "scTRP_clono"    # scTRP (+ TCR clonotype)
KEY_ONLY  = "scTRP_only"     # scTRP (transcription only)
KEY_XGB   = "xgboost"

# Top→bottom order in heatmap and (via reversed y_pos) strip+box
METHOD_ORDER = [KEY_CLONO, KEY_ONLY, KEY_XGB]

METHOD_LABELS = {
    KEY_CLONO: "scTRP",
    KEY_ONLY:  "scTRP\n(transcription only)",
    KEY_XGB:   "XGBoost",
}
METHOD_SHORT = {
    KEY_CLONO: "scTRP+TCR",
    KEY_ONLY:  "scTRP only",
    KEY_XGB:   "XGBoost",
}

_clono_pal = ["#d62728"] #sns.color_palette("OrRd", 6)[2:]
_only_pal  = sns.color_palette("PuBu", 6)[2:]
METHOD_PALETTE = {
    KEY_CLONO: _clono_pal[0],
    KEY_ONLY:  _only_pal[0],
    KEY_XGB:   "#888888",
}

METRICS = [
    ("ROC_AUC", "ROC-AUC",  0.4, 1.0),
    ("MCC",     "MCC",      0.0, 0.65),
    ("F1",      "F1-score", 0.4, 1.0),
    ("Gmean",   "G-mean",   0.4, 1.0),
]

# ── Metric helpers ────────────────────────────────────────────────────────────

def _jenks_thresh(scores):
    if HAS_JENKSPY and len(scores) >= 2:
        try:
            return jenkspy.jenks_breaks(list(scores), n_classes=2)[1]
        except Exception:
            pass
    return float(np.median(scores))


def _topn_precision(scores, labels, k):
    order    = np.argsort(scores)[::-1]
    actual_k = min(k, len(order))
    return float(np.array(labels)[order[:actual_k]].mean()) if actual_k > 0 else np.nan


def _topn_precision_two_score(score_i, score_ni, labels, k):
    """
    Rank cells by (score_i descending, score_ni descending as tiebreak),
    then compute precision among the top-k ranked cells.
    """
    score_i = np.asarray(score_i, dtype=float)
    score_ni = np.asarray(score_ni, dtype=float)
    labels = np.asarray(labels, dtype=int)
    if len(score_i) == 0:
        return np.nan
    # np.lexsort sorts by the LAST key first (primary), so score_i goes last
    order = np.lexsort((-score_ni, -score_i))
    top_idx = order[:k]
    return float(labels[top_idx].mean()) if len(top_idx) else np.nan

    
def _per_patient_topk_series(df, k, pat_col="sample", two_score=False):
    rows = {}
    for pat, g in df.groupby(pat_col):
        if two_score:
            rows[pat] = _topn_precision_two_score(
                g["score_i"].values.astype(float),
                g["score_ni"].values.astype(float),
                g["reactivity"].values.astype(int), k
            )
        else:
            rows[pat] = _topn_precision(
                g["score"].values.astype(float),
                g["reactivity"].values.astype(int), k
            )
    return pd.Series(rows)


def load_manascore_raw_scores(manascore_dir, study, pool_type, reference_df):
    """
    Load raw two-score MANAscore predictions for one study, restrict to
    barcodes present in `reference_df`, and attach the 'reactivity' label
    from that reference.

    Parameters
    ----------
    manascore_dir : str or Path
        Directory containing '{study}_sc_predictions.csv' /
        '{study}_clone_predictions.csv'.
    study : str
        Study name (e.g. 'chiffelle').
    pool_type : str
        'raw' -> uses '{study}_sc_predictions.csv'
        anything else (e.g. 'max') -> uses '{study}_clone_predictions.csv'
    reference_df : pd.DataFrame
        Any already-loaded, barcode-aligned cell-level dataframe for the
        same study/pool_type. Used to (a) restrict MANAscore to the common
        barcode set, and (b) supply the 'reactivity' label.

    Returns
    -------
    pd.DataFrame indexed by cell_barcode with columns
    ['score_i', 'score_ni', 'sample', 'reactivity'], or None if the source
    file is missing.
    """
    level = "sc" if pool_type == "raw" else "clone"
    fpath = Path(manascore_dir) / f"{study}_{level}_predictions.csv"
    if not fpath.exists():
        print(f"    [warn] MANAscore file not found: {fpath}")
        return None

    df = pd.read_csv(fpath, index_col=0)
    df.index.name = "cell_barcode"

    common_barcodes = df.index.intersection(reference_df.index)
    df = df.loc[common_barcodes].copy()
    df["reactivity"] = reference_df.loc[common_barcodes, "reactivity"].values

    keep_cols = [c for c in ["score_i", "score_ni", "sample", "reactivity"] if c in df.columns]
    return df[keep_cols]



    

# ── Wilcoxon helper ───────────────────────────────────────────────────────────

def _wilcoxon_stars(a, b, alternative="greater"):
    a, b = np.asarray(a, float), np.asarray(b, float)
    if len(a) < 3 or np.all(a == b):
        return np.nan, "ns"
    try:
        _, pval = ttest_rel(a, b, alternative=alternative)
        star = "***" if pval < 0.001 else "**" if pval < 0.01 else "*" if pval < 0.05 else "ns"
        return float(pval), star
    except Exception:
        return np.nan, ""


def _sig_bracket(ax, x1, x2, y, h=0.025, star="", fontsize=13,
                 color="#333333", lw=1.2, show_ns=False):
    if not star or (star == "ns" and not show_ns):
        return
    ax.plot([x1, x1, x2, x2], [y, y + h, y + h, y],
            lw=lw, color=color, clip_on=False)
    fw = "bold" if star != "ns" else "normal"
    ax.text((x1 + x2) / 2, y + h + 0.003, star,
            ha="center", va="bottom", fontsize=fontsize,
            color=color, fontweight=fw, clip_on=False)


# ── Data loading ──────────────────────────────────────────────────────────────

def load_all_patient_wise(in_root):
    csv = Path(in_root) / "all_patient_wise.csv"
    if not csv.exists():
        raise FileNotFoundError(
            f"{csv} not found — run compute_patient_metrics_all.py first"
        )
    df = pd.read_csv(csv)
    print(f"Loaded {len(df):,} rows  |  "
          f"pool_types: {sorted(df['pool_type'].unique())}")
    return df


def build_ablation_df(df_full, level):
    """
    Build ablation DataFrame for one level.
    Maps:
      scTRP_simclr (6 studies) + scTRP_simclr_rank_majority (Chiffelle) → KEY_CLONO
      scTRP_only   (6 studies) + scTRP_only_rank_majority   (Chiffelle) → KEY_ONLY
      xgboost (Chiffelle only, raw pool)                                → KEY_XGB
    """
    pool = POOL_CELL if level == "cell" else POOL_CLONO

    def _select(method_6, method_chif, new_key):
        six = df_full[
            (df_full["pool_type"] == pool) &
            (df_full["method"] == method_6) &
            (df_full["study"].isin(SIX_STUDIES))
        ].copy()
        chif = df_full[
            (df_full["pool_type"] == pool) &
            (df_full["method"] == method_chif) &
            (df_full["study"].isin(ENSEMBLE_STUDIES))
        ].copy()
        out = pd.concat([six, chif], ignore_index=True)
        out["method"] = new_key
        return out

    clono = _select("scTRP_simclr", f"scTRP_simclr_{ENSEMBLE}", KEY_CLONO)
    only  = _select("scTRP_only",   f"scTRP_only_{ENSEMBLE}",   KEY_ONLY)
    xgb   = df_full[
        (df_full["pool_type"] == pool) &
        (df_full["method"] == "xgboost")
    ].copy()
    xgb["method"] = KEY_XGB

    combined = pd.concat([clono, only, xgb], ignore_index=True)
    combined["level"] = level
    n = len(combined)
    print(f"  level={level}  rows={n:,}  "
          f"studies={sorted(combined['study'].unique())}")
    return combined


def load_cell_csvs_for_preck(in_root, pool_type):
    """Load all Chiffelle cell-level CSVs for one pool_type."""
    if pool_type == "raw":
        src = Path(in_root) / "raw_cell_scores"
    else:
        src = Path(in_root) / "pooled_scores" / pool_type
    if not src.exists():
        return {}
    result = {}
    for p in sorted(src.glob("*chiffelle*.csv")):
        stem = p.stem
        if stem.endswith("_chiffelle"):
            key = stem[:-len("_chiffelle")]
            try:
                result[key] = pd.read_csv(p, index_col=0)
            except Exception:
                pass
    return result


# ── Fig 1: Heatmap ────────────────────────────────────────────────────────────

def plot_heatmap(df_all, fig_dir, level):
    """Ablation trio × 7 studies. Dashed line before Chiffelle column."""
    fig_dir.mkdir(parents=True, exist_ok=True)
    mord = [m for m in METHOD_ORDER if m in df_all["method"].unique()]
    sord = [s for s in STUDY_ORDER if s in df_all["study"].unique()]

    for metric, title, vmin, vmax in METRICS:
        if metric not in df_all.columns or df_all[metric].isna().all():
            continue

        pivot = (df_all.groupby(["method", "study"])[metric].mean()
                 .unstack("study")
                 .reindex(index=mord, columns=sord))

        nrow, ncol = len(mord), len(sord)
        fig, ax = plt.subplots(figsize=(ncol * 1.9, nrow * 1.05),
                              constrained_layout=True)
        im = ax.imshow(pivot.values.astype(float), cmap="RdYlGn_r",
                       vmin=vmin, vmax=vmax, aspect="auto")

        for i in range(nrow):
            for j in range(ncol):
                val = pivot.values[i, j]
                if not np.isnan(val):
                    fw = "bold" if mord[i] == KEY_CLONO else "normal"
                    ax.text(j, i, f"{val:.2f}", ha="center", va="center",
                           fontsize=20, color="black", fontweight=fw)
                else:
                    ax.text(j, i, "–", ha="center", va="center",
                            fontsize=11, color="#888888")

        if "chiffelle" in sord:
            j_c = sord.index("chiffelle")
            ax.axvline(j_c - 0.5, color="#444", lw=2.0, ls="--")
        if KEY_XGB in mord:
            i_x = mord.index(KEY_XGB)
            ax.axhline(i_x - 0.5, color="#999", lw=1.2, ls=":")

        ax.set_xticks(range(ncol))
        ax.xaxis.tick_top()
        ax.xaxis.set_label_position("top")
        ax.set_xticklabels([STUDY_LABELS.get(s, s) for s in sord],
                            rotation=0, ha="center", fontsize=18)
        ax.set_yticks(range(nrow))
        ax.set_yticklabels([METHOD_LABELS.get(m, m) for m in mord], fontsize=18)
        ax.set_xlabel("Study", labelpad=8)
        ax.set_ylabel("Method", labelpad=8)
        ax.set_title(f"Ablation — {title}  [{level}]", pad=10, fontweight="bold")

        cbar = fig.colorbar(im, ax=ax, fraction=0.025, pad=0.02)
        cbar.ax.tick_params(labelsize=18)
        cbar.set_label(title, fontsize=20)
        for spine in ax.spines.values():
            spine.set_visible(False)

        fname = f"fig1_heatmap_{metric}_{level}"
        fig.savefig(fig_dir / f"{fname}.pdf")
        fig.savefig(fig_dir / f"{fname}.png", dpi=300)
        plt.show(fig)
        plt.close(fig)
        print(f"    saved {fname}")


# ── Fig 2: Grouped barplot per metric ─────────────────────────────────────────

def plot_grouped_barplot(df_all, fig_dir, level):
    fig_dir.mkdir(parents=True, exist_ok=True)
    mord        = [m for m in METHOD_ORDER if m in df_all["method"].unique()]
    print(mord)
    study_6     = [s for s in SIX_STUDIES if s in df_all["study"].unique()]
    study_chiff = ENSEMBLE_STUDIES if "moffitt" in df_all["study"].unique() else ["chiffelle"]
    study_chiff = [s for s in study_chiff if s in df_all["study"].unique()]

    n_meth = len(mord)
    #bar_w  = 0.80 / n_meth
    rng    = np.random.RandomState(42)

    KEY_S = KEY_CLONO
    KEY_O = KEY_ONLY
    IDX_S = mord.index(KEY_S) if KEY_S in mord else -1
    IDX_O = mord.index(KEY_O) if KEY_O in mord else -1
    IDX_X = mord.index(KEY_XGB) if KEY_XGB in mord else -1

    # Fixed bar width based on the WORST-CASE group size (all methods, i.e. the
    # chiffelle panel with XGBoost included) so bars render the same physical
    # thickness in both panels regardless of how many bars a given panel draws.
    BAR_W = 0.80 / n_meth

    PER_STUDY_W = 1.35  # inches per study column — same in both panels
    sig_rows = []
    for metric, metric_label, vmin, vmax in METRICS:
        if metric not in df_all.columns:
            continue

        fig, axes = plt.subplots(
            1, 2,
            figsize=((len(study_6) + len(study_chiff)) * PER_STUDY_W + 2.0, 7.2),
            gridspec_kw={"width_ratios": [max(len(study_6), 1),
                                          max(len(study_chiff), 1)]},
            sharey=True, constrained_layout=True,
        )

        
        for ax, studies_sub, panel_title in zip(
                axes,
                [study_6, study_chiff],
                ["6 leave-one-out test studies", "Independent test sets"]):
            if not studies_sub:
                ax.set_visible(False)
                continue

            x = np.arange(len(studies_sub))
            y_study_max = {s: vmin for s in studies_sub}
            pat_by_method = {m: df_all[df_all["method"] == m].set_index("sample")
                             for m in mord}
            # Methods actually drawn in THIS panel (XGB only appears in the
            # independent-test panel) — used to center bars correctly.
            active_methods = [m for m in mord
                               if not (m == KEY_XGB and studies_sub != study_chiff)]
            n_active = len(active_methods)

            # ── Draw bars ─────────────────────────────────────────────────────
            for i, method in enumerate(active_methods):
                offset = (i - n_active / 2 + 0.5) * BAR_W
                vals = [df_all[(df_all["study"] == s) &
                               (df_all["method"] == method)][metric].mean()
                        for s in studies_sub]
                bars = ax.bar(
                    x + offset, vals, width=BAR_W * 0.88,
                    color=METHOD_PALETTE[method], alpha=0.72,
                    edgecolor="white", linewidth=0.5,
                    label=METHOD_LABELS[method], zorder=3,
                )
                """
                for bar, v in zip(bars, vals):
                    if not np.isnan(v) and v > vmin:
                        ax.text(bar.get_x() + bar.get_width() / 2,
                                v + 0.006, f"{v:.3f}",
                                ha="center", va="bottom",
                                fontsize=16, rotation=90, fontweight="bold",
                                color="#111111", zorder=7)
                """

            # ── Per-patient scatter ───────────────────────────────────────────
            for i, method in enumerate(active_methods):
                offset = (i - n_active / 2 + 0.5) * BAR_W
                for j, study in enumerate(studies_sub):
                    ps = pat_by_method[method]
                    yvals = ps[ps["study"] == study][metric].dropna().values \
                        if "study" in ps.columns else ps[metric].dropna().values
                    if len(yvals) == 0:
                        continue
                    jitter = rng.uniform(-BAR_W * 0.27, BAR_W * 0.27, size=len(yvals))
                    ax.scatter(j + offset + jitter, yvals,
                               color=METHOD_PALETTE[method],
                               s=42, alpha=0.88,
                               edgecolors="black", linewidths=0.5, zorder=6)
                    y_study_max[study] = max(y_study_max[study],
                                             float(np.nanmax(yvals)))

            # ── Wilcoxon brackets ─────────────────────────────────────────────
            bracket_pairs = []

            if IDX_S >= 0 and IDX_O >= 0 and studies_sub != study_chiff:
                idx_a_local = active_methods.index(KEY_S)
                idx_b_local = active_methods.index(KEY_O)
                bracket_pairs.append((idx_a_local, idx_b_local, KEY_S, KEY_O,
                                      pat_by_method.get(KEY_S),
                                      pat_by_method.get(KEY_O), 0))

            if studies_sub == study_chiff:
                if IDX_S >= 0 and IDX_O >= 0:
                    idx_a_local = active_methods.index(KEY_S)
                    idx_b_local = active_methods.index(KEY_XGB)
                    bracket_pairs.append((idx_a_local, idx_b_local, KEY_S, KEY_XGB,
                                          pat_by_method.get(KEY_S),
                                          pat_by_method.get(KEY_O), 0))
                if IDX_S >= 0 and IDX_X >= 0:
                    bracket_pairs.append((IDX_S, IDX_X, KEY_S, KEY_XGB,
                                          pat_by_method.get(KEY_S),
                                          pat_by_method.get(KEY_XGB), 1))

            for (idx_a, idx_b, key_a, key_b, pa_full, pb_full, bracket_level) in bracket_pairs:
                if pa_full is None or pb_full is None:
                    continue
                for j, study in enumerate(studies_sub):
                    pa_s = pa_full[pa_full["study"] == study] if "study" in pa_full.columns else pa_full
                    pb_s = pb_full[pb_full["study"] == study] if "study" in pb_full.columns else pb_full
                    common = pa_s.index.intersection(pb_s.index)
                    if len(common) < 3:
                        continue
                    a = pa_s.loc[common, metric].dropna().values.astype(float)
                    b = pb_s.loc[common, metric].dropna().values.astype(float)
                    _, star = _wilcoxon_stars(a, b)

                    x_a = j + (idx_a - n_active / 2 + 0.5) * BAR_W
                    x_b = j + (idx_b - n_active / 2 + 0.5) * BAR_W
                    y_base = max(y_study_max[study], vmin + 0.02) + 0.08 + bracket_level * 0.08
                    _sig_bracket(ax, min(x_a, x_b), max(x_a, x_b),
                                 y_base, h=0.028, star=star, fontsize=16)

            # ── Save Wilcoxon significance to CSV ─────────────────────────────────────
            
            for (idx_a, idx_b, key_a, key_b, pa_full, pb_full, bracket_level) in bracket_pairs:
                if pa_full is None or pb_full is None:
                    continue
                for j, study in enumerate(studies_sub):
                    pa_s = pa_full[pa_full["study"] == study] if "study" in pa_full.columns else pa_full
                    pb_s = pb_full[pb_full["study"] == study] if "study" in pb_full.columns else pb_full
                    common = pa_s.index.intersection(pb_s.index)
                    if len(common) < 3:
                        continue
                    a = pa_s.loc[common, metric].dropna().values.astype(float)
                    b = pb_s.loc[common, metric].dropna().values.astype(float)
                    try:
                        from scipy.stats import wilcoxon as _wtest
                        from scipy.stats import ttest_rel
                        #_, pval = _wtest(a, b)
                        _, pval = ttest_rel(a, b)
                        pval_one, star = _wilcoxon_stars(a, b)
                    except Exception:
                        pval, pval_one, star = np.nan, np.nan, ""
                    sig_rows.append({
                        "metric":        metric,
                        "study":         study,
                        "level":         level,
                        "model_a":       key_a,
                        "model_b":       key_b,
                        "n_patients":    len(common),
                        "pvalue_two_sided": pval,
                        "pvalue_one_sided": pval_one,
                        "star":          star,
                        "test":          "paired T-test",
                    })
            
            
            
            ax.set_xticks(x)
            ax.set_xticklabels([STUDY_LABELS.get(s, s) for s in studies_sub],
                                ha="center", fontsize=16)
            ax.set_title(panel_title, fontsize=18, fontweight="bold", pad=20)
            ax.yaxis.grid(True, lw=0.6, color="#eeeeee", zorder=0)
            ax.set_xlim(-0.5, len(studies_sub) - 0.5)

        axes[0].set_ylabel(metric_label, fontsize=17, fontweight="bold")
        axes[0].set_ylim(max(vmin - 0.08, 0.0), min(vmax + 0.5, 1.3))
        axes[0].yaxis.set_major_formatter(mticker.FormatStrFormatter("%.2f"))
        axes[0].tick_params(axis="y", labelsize=16)

        handles, lbs = [], []
        for ax in axes:
            if ax.get_visible():
                for h_, l in zip(*ax.get_legend_handles_labels()):
                    if l not in lbs:
                        handles.append(h_); lbs.append(l)
            
        fig.legend(handles, lbs, loc="lower center", ncol=min(n_meth, 5),
                   frameon=False, fontsize=16, bbox_to_anchor=(0.5, -0.20),
                   title=f"Model variant [{level}-level]", title_fontsize=16)

        fname = f"fig2_barplot_{metric}_{level}"
        fig.savefig(fig_dir / f"{fname}.pdf", bbox_inches="tight")
        fig.savefig(fig_dir / f"{fname}.png", dpi=300, bbox_inches="tight")
        plt.show()
        plt.close(fig)
        print(f"    saved {fname}")
    
    if sig_rows:
        df_sig = pd.DataFrame(sig_rows)
        sig_fname = fig_dir / f"ablation_barplot_t_test_{level}.csv"
        df_sig.to_csv(sig_fname, index=False, float_format="%.6f")
        print(f"    saved {sig_fname}")

    

# ── Fig 3: Δ bar chart ────────────────────────────────────────────────────────

def plot_delta_bars(df_all, fig_dir, level):
    """Δ = KEY_CLONO − KEY_ONLY per study. Red = clonotype helps, blue = hurts."""
    fig_dir.mkdir(parents=True, exist_ok=True)
    sord = [s for s in STUDY_ORDER if s in df_all["study"].unique()]

    delta_rows = []
    for study in sord:
        for metric, *_ in METRICS:
            if metric not in df_all.columns:
                continue
            sub = df_all[df_all["study"] == study]
            v_c = sub[sub["method"] == KEY_CLONO][metric].mean()
            v_o = sub[sub["method"] == KEY_ONLY][metric].mean()
            delta_rows.append({"study": study, "metric": metric,
                                "clono": v_c, "only": v_o, "delta": v_c - v_o})
    df_delta = pd.DataFrame(delta_rows)
    if df_delta.empty:
        print("    [skip] no data for delta plot")
        return

    n_met = len(METRICS)
    fig, axes = plt.subplots(1, n_met, figsize=(n_met * 3.6 + 0.8, 4.5),
                              sharey=False, constrained_layout=True)
    if n_met == 1:
        axes = [axes]

    for ax, (metric, metric_label, *_) in zip(axes, METRICS):
        sub    = df_delta[df_delta["metric"] == metric].set_index("study").reindex(sord)
        deltas = sub["delta"].values.astype(float)
        colors = ["#d62728" if d > 0 else "#4878d0" for d in deltas]
        x      = np.arange(len(sord))

        bars = ax.bar(x, deltas, color=colors, alpha=0.85,
                      edgecolor="white", linewidth=0.5, zorder=3)
        ax.axhline(0, color="black", lw=1.2, zorder=2)
        ax.yaxis.grid(True, lw=0.6, color="#eeeeee", zorder=0)

        for bar, v in zip(bars, deltas):
            if not np.isnan(v):
                ax.text(bar.get_x() + bar.get_width() / 2,
                        v + (0.004 if v >= 0 else -0.004),
                        f"{v:+.3f}", ha="center",
                        va="bottom" if v >= 0 else "top",
                        fontsize=11, fontweight="bold",
                        color="#d62728" if v > 0 else "#4878d0")

        ax.set_xticks(x)
        ax.set_xticklabels([STUDY_LABELS.get(s, s) for s in sord],
                            ha="center", fontsize=12)
        ax.set_title(metric_label, fontsize=16, fontweight="bold", pad=6)
        if ax is axes[0]:
            ax.set_ylabel("Δ (+ clonotype − only)", fontsize=15, fontweight="bold")
        ax.tick_params(axis="y", labelsize=13)
        if "chiffelle" in sord:
            j = sord.index("chiffelle")
            ax.axvline(j - 0.5, color="#888", lw=1.5, ls="--", alpha=0.7)

    fig.legend(
        [mpatches.Patch(color="#d62728", alpha=0.85),
         mpatches.Patch(color="#4878d0", alpha=0.85)],
        ["+TCR clonotype helps", "+TCR clonotype hurts"],
        loc="upper right", frameon=False, fontsize=13, bbox_to_anchor=(1.0, 1.02)
    )
    fig.suptitle(
        f"Δ scTRP: (+ TCR clonotype) − (transcription only)  [{level}]",
        fontsize=16, fontweight="bold"
    )

    fname = f"fig3_delta_{level}"
    #fig.savefig(fig_dir / f"{fname}.pdf", bbox_inches="tight")
    #fig.savefig(fig_dir / f"{fname}.png", dpi=300, bbox_inches="tight")
    plt.show(fig)
    plt.close(fig)
    print(f"    saved {fname}")


# ── Fig 4: Per-patient strip+box on Chiffelle ─────────────────────────────────

def plot_per_patient_chiffelle(df_all, fig_dir, level):
    """
    Strip+box for Chiffelle patients only: KEY_CLONO vs KEY_ONLY.
    Y-axis top→bottom: KEY_CLONO (+ clonotype) at top, KEY_ONLY at bottom.
    Paired Wilcoxon (one-sided) annotated per metric panel.
    """
    fig_dir.mkdir(parents=True, exist_ok=True)
    mord = [m for m in [KEY_CLONO, KEY_ONLY] if m in df_all["method"].unique()]
    if len(mord) < 2:
        print("    [skip] per-patient plot: need both KEY_CLONO and KEY_ONLY")
        return

    df_chif = df_all[df_all["study"] == "chiffelle"]
    if df_chif.empty:
        print("    [skip] no Chiffelle data")
        return

    n_meth  = len(mord)
    metrics_present = [(m, ml, xn, xx) for m, ml, xn, xx in METRICS
                       if m in df_chif.columns]
    n_met   = len(metrics_present)

    # Reversed y positions: KEY_CLONO at top (highest y)
    y_pos = {m: n_meth - 1 - i for i, m in enumerate(mord)}

    fig, axes = plt.subplots(1, n_met, figsize=(n_met * 3.8, 4.2),
                              constrained_layout=True)
    if n_met == 1:
        axes = [axes]

    rng = np.random.RandomState(42)

    for ax, (metric, metric_label, xmin, xmax) in zip(axes, metrics_present):
        sub_c = (df_chif[df_chif["method"] == KEY_CLONO]
                 .dropna(subset=[metric]).set_index("sample"))
        sub_o = (df_chif[df_chif["method"] == KEY_ONLY]
                 .dropna(subset=[metric]).set_index("sample"))

        for method in mord:
            msub  = df_chif[df_chif["method"] == method][metric].dropna().values
            if len(msub) == 0:
                continue
            yp    = y_pos[method]
            color = METHOD_PALETTE[method]

            if len(msub) >= 2:
                q1, q2, q3 = np.percentile(msub, [25, 50, 75])
                iqr = q3 - q1
                lo, hi = max(msub.min(), q1 - 1.5 * iqr), min(msub.max(), q3 + 1.5 * iqr)
                ax.barh(yp, q3 - q1, left=q1, height=0.50,
                        color=color, alpha=0.28, zorder=2)
                ax.vlines(q2, yp - 0.25, yp + 0.25, color=color, lw=3.0, zorder=3)
                ax.hlines(yp, lo, hi, color=color, lw=1.2, alpha=0.65, zorder=2)

            jit = rng.uniform(-0.14, 0.14, size=len(msub))
            ax.scatter(msub, yp + jit, color=color, s=64,
                       alpha=0.90, edgecolors="white", linewidths=0.6, zorder=4)

        # Paired Wilcoxon
        common = sub_c.index.intersection(sub_o.index)
        if len(common) >= 3:
            a = sub_c.loc[common, metric].values.astype(float)
            b = sub_o.loc[common, metric].values.astype(float)
            pval, star = _wilcoxon_stars(a, b)
            print(star)
            if not np.isnan(pval):
                ax.text(0.97, 0.97, f"p = {pval:.3f}  {star}",
                        transform=ax.transAxes, ha="right", va="top", fontsize=13,
                        fontweight="bold",
                        bbox=dict(boxstyle="round,pad=0.3", fc="white",
                                  ec="#999999", alpha=0.85))

        ax.set_xlim(xmin, xmax + 0.05)
        ax.set_ylim(-0.5, n_meth - 0.5)
        ax.axvline(0.5, color="lightgray", lw=0.8, ls="--", zorder=1)
        ax.set_title(metric_label, fontsize=15, fontweight="bold", pad=6)
        ax.set_xlabel(metric_label, fontsize=13)
        ax.set_yticks(range(n_meth))

        if ax is axes[0]:
            ax.set_yticklabels(
                [METHOD_LABELS.get(m, m) for m in reversed(mord)], fontsize=13
            )
            ax.set_ylabel("Method", labelpad=8, fontsize=14)
        else:
            ax.set_yticklabels([])

    fig.suptitle(f"Chiffelle — per-patient: +TCR clonotype vs transcription only  [{level}]",
                 fontsize=14, fontweight="bold")

    fname = f"fig4_per_patient_chiffelle_{level}"
    fig.savefig(fig_dir / f"{fname}.pdf")
    fig.savefig(fig_dir / f"{fname}.png", dpi=300)
    plt.show(fig)
    plt.close(fig)
    print(f"    saved {fname}")


# ── Fig 5: Precision@k bar chart ─────────────────────────────────────────────


def load_cell_csvs_for_preck(in_root, pool_type, study):
    """Load all cell-level CSVs for one (study, pool_type) combination."""
    if pool_type == "raw":
        src = Path(in_root) / "raw_cell_scores"
    else:
        src = Path(in_root) / "pooled_scores" / pool_type
    if not src.exists():
        return {}
    result = {}
    suffix = f"_{study}"
    for p in sorted(src.glob(f"*{study}*.csv")):
        stem = p.stem
        if stem.endswith(suffix):
            key = stem[:-len(suffix)]
            try:
                df_p = pd.read_csv(p, index_col=0)
                
                
                result[key] = df_p
            except Exception:
                pass
    return result


'''
def plot_precision_k(cell_data_raw, cell_data_pool, fig_dir, level,
                      study, manascore_dir=None,
                      k_vals=(10, 20, 30, 40, 50)):
    """
    Grouped bar: precision@k for a given study.
    clono-level: cell_data_pool (max pool)
    cell-level : cell_data_raw (raw)
    Reference for Wilcoxon = scTRP_simclr (chiffelle: scTRP_simclr_rank_majority).

    MANAscore special case: if `manascore_dir` is given, the
    'benchmark_Manascore' entry is overridden with raw two-score predictions
    ('score_i', 'score_ni') loaded from
    '{manascore_dir}/{study}_sc_predictions.csv' (raw) or
    '{manascore_dir}/{study}_clone_predictions.csv' (pooled). Rows are
    restricted to barcodes already present in the other methods' loaded
    data, and precision@k ranks by score_i descending with score_ni
    descending as a tiebreaker, instead of a single 'score' column.

    Results are written to fig_dir / study / ...
    """
    from scipy.stats import wilcoxon as _wtest

    out_dir = fig_dir / study
    out_dir.mkdir(parents=True, exist_ok=True)

    pool_type = "max" if level == "clonotype" else "raw"
    cell_data = cell_data_pool if level == "clonotype" else cell_data_raw

    # scTRP key naming differs for chiffelle (assembled/rank-majority results)
    # vs. all other studies (plain scTRP variants).
    if study == "chiffelle":
        sctrp_simclr_raw = f"scTRP_simclr_{ENSEMBLE}"
        sctrp_only_raw    = f"scTRP_only_{ENSEMBLE}"
    else:
        sctrp_simclr_raw = "scTRP_simclr"
        sctrp_only_raw   = "scTRP_only"

    key_map = {
        sctrp_simclr_raw:           KEY_CLONO,
        sctrp_only_raw:             KEY_ONLY,
        "xgboost":                  KEY_XGB,
        "benchmark_Seurat_Caushi":  "benchmark_Seurat_Caushi",
        "benchmark_Hanada":         "benchmark_Hanada",
        "benchmark_AUCell_Lowery":  "benchmark_AUCell_Lowery",
        "benchmark_UCell_Meng":     "benchmark_UCell_Meng",
        "benchmark_UCell_Oliveira": "benchmark_UCell_Oliveira",
        "benchmark_Zheng":          "benchmark_Zheng",
        "benchmark_Manascore":      "benchmark_Manascore",
        "benchmark_Trace":          "benchmark_Trace",
    }
    bench_display_keys = [
        "benchmark_Seurat_Caushi", "benchmark_Hanada", "benchmark_AUCell_Lowery",
        "benchmark_UCell_Meng", "benchmark_UCell_Oliveira", "benchmark_Zheng",
        "benchmark_Manascore", "benchmark_Trace",
    ]
    bench_labels = {
        "benchmark_Seurat_Caushi":  "Caushi et al.\n(Seurat)",
        "benchmark_Hanada":         "Hanada et al.",
        "benchmark_AUCell_Lowery":  "NeoTCR8\n(AUCell)",
        "benchmark_UCell_Meng":     "TR30\n(UCell)",
        "benchmark_UCell_Oliveira": "Oliveira\n(UCell)",
        "benchmark_Zheng":          "Zheng et al.",
        "benchmark_Manascore":      "MANAscore",
        "benchmark_Trace":          "TRACE",
    }
    bench_pal = sns.color_palette('tab10', n_colors=len(bench_display_keys))
    display_palette = {k: c for k, c in zip(bench_display_keys, bench_pal)}
    display_palette.update(METHOD_PALETTE)

    # Build unified {display_key: df}
    unified = {}
    for raw_key, df in cell_data.items():
        dk = key_map.get(raw_key, raw_key)
        if dk not in unified:
            unified[dk] = df

    # --- MANAscore override: load raw two-score predictions ---------------
    manascore_two_score = None
    if manascore_dir is not None and unified:
        reference_df = next(iter(unified.values()))
        manascore_two_score = load_manascore_raw_scores(
            manascore_dir, study, pool_type, reference_df
        )
        if manascore_two_score is not None and len(manascore_two_score):
            unified["benchmark_Manascore"] = manascore_two_score
        else:
            print(f"    [warn] MANAscore override failed/empty for {study}; "
                  "keeping original entry if present")

    method_order_prec = [k for k in bench_display_keys if k in unified] + \
                        [m for m in [KEY_CLONO] if m in unified]
    if not method_order_prec:
        print(f"    [skip] precision@k ({study}): no matching keys in {list(unified.keys())}")
        return

    baseline = (next(iter(unified.values()))["reactivity"].astype(int).mean()
                if unified else np.nan)

    def _is_manascore_two_score(dk):
        return dk == "benchmark_Manascore" and manascore_two_score is not None \
            and "score_i" in unified[dk].columns and "score_ni" in unified[dk].columns

    def _required_cols(dk):
        return ["reactivity", "score_i", "score_ni"] if _is_manascore_two_score(dk) \
            else ["reactivity", "score"]

    def _precision_for_group(dk, g, k):
        if _is_manascore_two_score(dk):
            return _topn_precision_two_score(
                g["score_i"].values.astype(float),
                g["score_ni"].values.astype(float),
                g["reactivity"].values.astype(int), k
            )
        return _topn_precision(
            g["score"].values.astype(float),
            g["reactivity"].values.astype(int), k
        )

    # Overall precision@k per method
    prec_rows = {}
    for dk in method_order_prec:
        g = unified[dk].dropna(subset=_required_cols(dk))
        if level == "clonotype":
            g.drop_duplicates(subset='sample_clone_id', inplace=True)
        prec_rows[dk] = {k: _precision_for_group(dk, g, k) for k in k_vals}
    bar_data = pd.DataFrame(prec_rows, index=list(k_vals)).T
    bar_data_ = bar_data.copy()
    bar_data_['level'] = f'{level}'
    bar_data_['study'] = study
    bar_data_fname = out_dir / f"fig5_precision_k_{level}.csv"
    bar_data_.to_csv(bar_data_fname, index=True, float_format="%.6f")
    print(f"    saved {bar_data_fname}")

    # ── Per-patient precision@k ───────────────────────────────────────────
    pat_rows = []
    for dk in method_order_prec:
        df_m = unified[dk].dropna(subset=_required_cols(dk))
        if level == "clonotype":
            df_m.drop_duplicates(subset='sample_clone_id', inplace=True)
        for patient, df_p in df_m.groupby("sample"):
            for k in k_vals:
                prec = _precision_for_group(dk, df_p, k)
                pat_rows.append({
                    "method":   dk,
                    "sample":   patient,
                    "k":        k,
                    "precision": prec,
                    "level":    level,
                    "study":    study,
                })

    df_pat_prec = pd.DataFrame(pat_rows)
    pat_prec_fname = out_dir / f"fig5_precision_k_per_patient_{level}.csv"
    df_pat_prec.to_csv(pat_prec_fname, index=False, float_format="%.6f")
    print(f"    saved {pat_prec_fname}")

    # Per-patient Wilcoxon: ref = KEY_CLONO
    ref_key = KEY_CLONO if KEY_CLONO in unified else None
    stars = defaultdict(dict)
    if ref_key:
        df_ref = unified[ref_key].dropna(subset=["score", "reactivity"])
        if level == "clonotype":
            df_ref.drop_duplicates(subset='sample_clone_id', inplace=True)
        for bk in bench_display_keys:
            if bk not in unified:
                continue
            before = unified[bk]
            n_before = len(before)
            
            
            df_b = unified[bk].dropna(subset=["score", "reactivity"])
            if level == "clonotype":
                df_b.drop_duplicates(subset='sample_clone_id', inplace=True)
            n_dropped = n_before - len(df_b)
            print(f"    [{study}] {bk}: dropped {n_dropped} of {n_before} rows")
            for k in k_vals:
                p_ref = _per_patient_topk_series(df_ref, k)
                p_b   = _per_patient_topk_series(df_b, k)
                common = p_ref.index.intersection(p_b.index)
                if len(common) < 3:
                    stars[bk][k] = ""
                    continue
                a = p_ref.loc[common].values.astype(float)
                b = p_b.loc[common].values.astype(float)
                try:
                    _, pval = _wtest(a, b, alternative="greater")
                    stars[bk][k] = ("***" if pval < 0.001 else
                                    "**"  if pval < 0.01  else
                                    "*"   if pval < 0.05  else "ns")
                except Exception:
                    stars[bk][k] = ""

    # ── Save stars and p-values to CSV ───────────────────────────────────────
    pval_rows = []
    if ref_key:
        df_ref = unified[ref_key].dropna(subset=["score", "reactivity"])
        
        if level == "clonotype":
            df_ref.drop_duplicates(subset='sample_clone_id', inplace=True)
        for bk in bench_display_keys:
            if bk not in unified:
                continue

            df_b = unified[bk].dropna(subset=["score", "reactivity"])
            if level == "clonotype":
                df_b.drop_duplicates(subset='sample_clone_id', inplace=True)
            for k in k_vals:
                p_ref = _per_patient_topk_series(df_ref, k)
                p_b   = _per_patient_topk_series(df_b, k)
                common = p_ref.index.intersection(p_b.index)
                if len(common) < 3:
                    pval_rows.append({
                        "benchmark":   bk,
                        "benchmark_label": bench_labels.get(bk, bk),
                        "k":           k,
                        "n_patients":  len(common),
                        "pvalue":      np.nan,
                        "star":        "",
                        "reference":   ref_key,
                        "test":        "paired Wilcoxon (greater)",
                        "level":       level,
                        "study":       study,
                    })
                    continue
                a = p_ref.loc[common].values.astype(float)
                b = p_b.loc[common].values.astype(float)
                try:
                    _, pval = _wtest(a, b, alternative="greater")
                    star = ("***" if pval < 0.001 else
                            "**"  if pval < 0.01  else
                            "*"   if pval < 0.05  else "ns")
                except Exception:
                    pval, star = np.nan, ""
                pval_rows.append({
                    "benchmark":       bk,
                    "benchmark_label": bench_labels.get(bk, bk),
                    "k":               k,
                    "n_patients":      len(common),
                    "pvalue":          pval,
                    "star":            star,
                    "reference":       ref_key,
                    "test":            "paired Wilcoxon (greater)",
                    "level":           level,
                    "study":           study,
                })

    df_pval = pd.DataFrame(pval_rows)
    pval_fname = out_dir / f"fig5_precision_k_wilcoxon_{level}.csv"
    df_pval.to_csv(pval_fname, index=False, float_format="%.6f")
    print(f"    saved {pval_fname}")

    # Plot
    n_methods = len(method_order_prec)
    n_k       = len(k_vals)
    x         = np.arange(n_k)
    bar_w     = 0.80 / n_methods
    sctrp_set = {KEY_CLONO, KEY_ONLY, KEY_XGB}

    fig, ax = plt.subplots(figsize=(max(8, n_k * 1.6 + 2), 5.5),
                           constrained_layout=True)

    all_labels = dict(bench_labels)
    all_labels.update({k: METHOD_LABELS[k].replace("\n", " ")
                       for k in [KEY_CLONO, KEY_ONLY, KEY_XGB] if k in METHOD_LABELS})

    for i, dk in enumerate(method_order_prec):
        offset   = (i - n_methods / 2 + 0.5) * bar_w
        vals     = [bar_data.loc[dk, k] if k in bar_data.columns else np.nan
                    for k in k_vals]
        is_sctrp = dk in sctrp_set
        bars = ax.bar(
            x + offset, vals, width=bar_w * 0.92,
            color=display_palette.get(dk, "#aaa"),
            alpha=1.0 if is_sctrp else 0.72,
            edgecolor="white", linewidth=0.4,
            label=all_labels.get(dk, dk),
            zorder=3 if is_sctrp else 2,
        )
        if is_sctrp:
            for bar, v in zip(bars, vals):
                if not np.isnan(v):
                    ax.text(bar.get_x() + bar.get_width() / 2, v + 0.012,
                            f"{v:.0%}", ha="center", va="bottom",
                            fontsize=10, fontweight="bold",
                            color=display_palette.get(dk, "#333"))
        else:
            for bar, v, k in zip(bars, vals, k_vals):
                star = stars[dk].get(k, "")
                if star and star != "ns" and not np.isnan(v):
                    ax.text(bar.get_x() + bar.get_width() / 2, v + 0.006,
                            star, ha="center", va="bottom", fontsize=14,
                            color=display_palette.get(dk, "#555"))

    if not np.isnan(baseline):
        ax.axhline(baseline, color="#888888", lw=1.1, ls="--",
                   label=f"Baseline ({baseline:.0%})")

    ax.set_xticks(x)
    ax.set_xticklabels([f"Top-{k}" for k in k_vals])
    ax.set_ylabel("Precision (fraction reactive)", fontsize=16)
    ax.set_title(f"Precision@k — {study}  [{level}]", pad=15, fontweight="bold", y=1.15)
    ax.set_ylim(0, ax.get_ylim()[1] * 1.05)
    ax.yaxis.set_major_formatter(mticker.PercentFormatter(xmax=1, decimals=0))
    ax.yaxis.grid(True, lw=0.5, color="#eeeeee", zorder=0)

    handles, lbs = ax.get_legend_handles_labels()
    ax.legend(handles, lbs, frameon=False, fontsize=12, loc="upper right", ncol=5, bbox_to_anchor=(1.0, 1.20))
    ax.annotate(
        "Significance: paired Wilcoxon, scTRP > benchmark (one-sided); * shows p-value < 0.05",
        xy=(0.01, -0.12), xycoords="axes fraction",
        fontsize=14, color="gray", va="bottom"
    )

    fname = f"fig5_precision_k_{level}"
    fig.savefig(out_dir / f"{fname}.pdf")
    fig.savefig(out_dir / f"{fname}.png", dpi=300)
    plt.show(fig)
    plt.close(fig)
    print(f"    saved {fname}")
'''
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from matplotlib.patches import Patch
from collections import defaultdict
from scipy.stats import wilcoxon as _wtest
from scipy.stats import ttest_rel
from scipy.stats import ttest_ind



BENCH_DISPLAY_KEYS = [
    "benchmark_Seurat_Caushi", "benchmark_Hanada", "benchmark_AUCell_Lowery",
    "benchmark_UCell_Meng", "benchmark_UCell_Oliveira", "benchmark_Zheng",
    "benchmark_Manascore", "benchmark_Trace",
]
BENCH_LABELS = {
    "benchmark_Seurat_Caushi":  "Caushi et al.\n(Seurat)",
    "benchmark_Hanada":         "Hanada et al.",
    "benchmark_AUCell_Lowery":  "NeoTCR8\n(AUCell)",
    "benchmark_UCell_Meng":     "TR30\n(UCell)",
    "benchmark_UCell_Oliveira": "Oliveira et al.\n(Seurat)",
    "benchmark_Zheng":          "Zheng et al.",
    "benchmark_Manascore":      "MANAscore",
    "benchmark_Trace":          "TRACE",
}

# Studies whose scTRP entries are ensemble-assembled (keys carry the mode suffix)
ENSEMBLE_STUDIES = ["chiffelle", "moffitt"]

# Publication font sizes (single place to retune)
FS = {"title": 16, "axlabel": 18, "tick": 12, "legend": 12,
      "star": 14, "value": 11, "note": 12}


def _as_int_reactivity(s):
    """Normalize reactivity to nullable Int64 so nunique() can't be fooled by mixed dtypes."""
    return (pd.to_numeric(s.astype(str).str.strip(), errors="coerce")
              .round().astype("Int64"))


def _study_subset(cell_data, study):
    """Accept {(key, study): df} or an already-per-study {key: df}."""
    first = next(iter(cell_data), None)
    if isinstance(first, tuple):
        return {k: df for (k, s), df in cell_data.items() if s == study}
    return dict(cell_data)


def _sctrp_raw_keys(study, ensemble=ENSEMBLE, ensemble_studies=ENSEMBLE_STUDIES):
    """
    X's conditional naming: ensemble studies carry the mode suffix on scTRP keys,
    all other studies use the plain variants.
    """
    if study in ensemble_studies:
        return f"scTRP_simclr_{ensemble}", f"scTRP_only_{ensemble}"
    return "scTRP_simclr", "scTRP_only"


def _prep_precision_ctx(cell_data, study, pool_type, ensemble=ENSEMBLE,
                        ensemble_studies=ENSEMBLE_STUDIES, manascore_dir=None,
                        removeNoPositivePatient=False,
                        sctrp_display_keys=(None,), k_vals=(10, 20, 30, 40, 50)):
    """
    Build everything both figures need for ONE study: unified {display_key: df},
    method order, palette, labels, baseline, reference key, study-level precision@k,
    and the per-(method, k) per-patient precision series.

    The per-patient series is computed ONCE here and reused by the plot, the CSVs,
    and the statistics, so the three can never drift apart.
    """
    sub = _study_subset(cell_data, study)
    if not sub:
        return None

    suffix = ""
    if removeNoPositivePatient:
        suffix = "_rmNopositivePatient"
        _ref = next(iter(sub.values()))
        _nuniq = _as_int_reactivity(_ref["reactivity"]).groupby(_ref["sample"]).nunique()
        keep = set(_nuniq[_nuniq > 1].index)
        dropped = sorted(set(_nuniq.index) - keep)
        if dropped:
            print(f"    [{study}] dropping {len(dropped)} single-class patient(s): {dropped}")
        if not keep:
            print(f"    [skip] {study}: no patients with both classes")
            return None
        sub = {k: df[df["sample"].isin(keep)].copy() for k, df in sub.items()}

    # ── X's key_map, with study-conditional scTRP naming ─────────────────────
    sctrp_simclr_raw, sctrp_only_raw = _sctrp_raw_keys(study, ensemble, ensemble_studies)
    key_map = {
        sctrp_simclr_raw: KEY_CLONO,
        sctrp_only_raw:   KEY_ONLY,
        "xgboost":        KEY_XGB,
        **{b: b for b in BENCH_DISPLAY_KEYS},
    }
    unified = {}
    for raw_key, df in sub.items():
        #print('===='*4)
        #print('check sub key:')
        #print(raw_key)
        dk = key_map.get(raw_key)
        if dk is not None and dk not in unified:
            unified[dk] = df
    if not unified:
        print(f"    [skip] {study}: no mapped methods (expected {sctrp_simclr_raw!r}); "
              f"saw {sorted(sub)[:6]}...")
        return None

    # ── MANAscore override (single-score frame) ──────────────────────────────
    if manascore_dir is not None:
        ms = load_manascore_raw_scores(manascore_dir, study, pool_type,
                                       next(iter(unified.values())))
        if ms is not None and len(ms) and "score" in ms.columns:
            unified["benchmark_Manascore"] = ms
        else:
            print(f"    [warn] {study}: MANAscore override failed/empty; keeping original")

    sctrp_keys = [k for k in sctrp_display_keys if k is not None]
    print(f'check sctrp_keys: {sctrp_keys}')
    method_order = ([m for m in sctrp_keys if m in unified] + 
                    [b for b in BENCH_DISPLAY_KEYS if b in unified])
    if not method_order:
        print(f"    [skip] {study}: no methods to plot")
        return None

    # ── Colors + labels (X's scheme) ─────────────────────────────────────────
    bench_pal = sns.color_palette("tab10", n_colors=len(BENCH_DISPLAY_KEYS))
    palette = {k: c for k, c in zip(BENCH_DISPLAY_KEYS, bench_pal)}
    palette.update(METHOD_PALETTE)
    labels = dict(BENCH_LABELS)
    labels.update({k: METHOD_LABELS[k]#.replace("\n", " ")
                   for k in (KEY_CLONO, KEY_ONLY, KEY_XGB) if k in METHOD_LABELS})

    # ── Clean each method's frame (non-mutating; X used inplace=True) ────────
    prepped = {}
    for dk in method_order:
        d = unified[dk].copy().dropna(subset=["score", "reactivity"])
        if pool_type != "raw" and "sample_clone_id" in d.columns:
            d = d.drop_duplicates(subset="sample_clone_id")
        prepped[dk] = d

    def _prec(g, k):
        return _topn_precision(g["score"].values.astype(float),
                               g["reactivity"].values.astype(int), k)

    baseline = _as_int_reactivity(
        next(iter(prepped.values()))["reactivity"]).astype(float).mean()

    study_prec  = {dk: {k: _prec(prepped[dk], k) for k in k_vals} for dk in method_order}
    per_patient = {
        (dk, k): pd.Series({pat: _prec(dfp, k)
                            for pat, dfp in prepped[dk].groupby("sample")}, dtype=float)
        for dk in method_order for k in k_vals
    }
    n_pat = max((len(s) for s in per_patient.values()), default=0)

    return dict(unified=unified, prepped=prepped, method_order=method_order,
                palette=palette, labels=labels, baseline=baseline, suffix=suffix,
                ref_key=(KEY_CLONO if KEY_CLONO in prepped else None),
                study_prec=study_prec, per_patient=per_patient, n_patients=n_pat,
                sctrp_set={KEY_CLONO, KEY_ONLY, KEY_XGB}, k_vals=tuple(k_vals),
                pool_type=pool_type, study=study,
                level=("clonotype" if pool_type != "raw" else "cell"))


def _star_from_p(p):
    return "***" if p < 0.001 else "**" if p < 0.01 else "*" if p < 0.05 else "ns"


def _paired_stats(a_s, b_s, alternative, min_pairs=3):
    """Per-patient paired t-test on two precision@k series. Patients = replicates."""
    common = a_s.index.intersection(b_s.index)
    out = {"n_patients": len(common), "statistic": np.nan, "pvalue": np.nan,
           "star": "", "ref_mean": np.nan, "method_mean": np.nan, "mean_diff": np.nan}
    if not len(common):
        return out
    a, b = a_s.loc[common].values.astype(float), b_s.loc[common].values.astype(float)
    out.update(ref_mean=float(a.mean()), method_mean=float(b.mean()),
               mean_diff=float((a - b).mean()))
    if len(common) < min_pairs:
        return out
    if np.all(a - b == 0):
        out["star"] = "ns"
        return out
    try:
        st, pv = ttest_rel(a, b, alternative=alternative)
        out.update(statistic=float(st), pvalue=float(pv), star=_star_from_p(pv))
    except Exception:
        pass
    return out


def _save_ctx_tables(ctx, out_dir, base, alternative):
    """Write per-patient values, study-level values, and t-test statistics."""
    meta = {"study": ctx["study"], "level": ctx["level"], "pool_type": ctx["pool_type"]}

    rows = [{**meta, "method": dk,
             "method_label": ctx["labels"].get(dk, dk).replace("\n", " "),
             "k": int(k), "sample": pat, "precision": float(v)}
            for (dk, k), s in ctx["per_patient"].items() for pat, v in s.items()]
    pd.DataFrame(rows).to_csv(out_dir / f"{base}_per_patient.csv",
                              index=False, float_format="%.6f")

    srows = [{**meta, "method": dk,
              "method_label": ctx["labels"].get(dk, dk).replace("\n", " "),
              "k": int(k), "study_precision": float(v)}
             for dk, d in ctx["study_prec"].items() for k, v in d.items()]
    pd.DataFrame(srows).to_csv(out_dir / f"{base}_studylevel.csv",
                               index=False, float_format="%.6f")

    ref = ctx["ref_key"]
    if ref is None:
        return
    wrows = []
    for dk in ctx["method_order"]:
        if dk == ref:
            continue
        for k in ctx["k_vals"]:
            st = _paired_stats(ctx["per_patient"][(ref, k)],
                               ctx["per_patient"][(dk, k)], alternative)
            wrows.append({**meta, "reference": ref, "method": dk,
                          "method_label": ctx["labels"].get(dk, dk).replace("\n", " "),
                          "k": int(k), "test": f"paired t-test ({alternative})", **st})
    pd.DataFrame(wrows).to_csv(out_dir / f"{base}_t_test.csv",
                               index=False, float_format="%.6f")

def compute_max_n_m(cell_data, studies, pool_type, ensemble=ENSEMBLE,
                    ensemble_studies=ENSEMBLE_STUDIES, manascore_dir=None,
                    removeNoPositivePatient=False,
                    sctrp_display_keys=(KEY_CLONO,), k_vals=(10, 20, 30, 40, 50)):
    """Scan all studies once to find the largest method_order, so every
    per-study figure can share one consistent figure width."""
    n_m_list = []
    for study in studies:
        ctx = _prep_precision_ctx(cell_data, study, pool_type, ensemble, ensemble_studies,
                                  manascore_dir=manascore_dir,
                                  removeNoPositivePatient=removeNoPositivePatient,
                                  sctrp_display_keys=sctrp_display_keys, k_vals=k_vals)
        if ctx is not None:
            n_m_list.append(len(ctx["method_order"]))
    return max(n_m_list) if n_m_list else 1


def plot_study_precision_k(cell_data, fig_dir, study, pool_type,
                           k_vals=(10, 15, 20), ensemble=ENSEMBLE,
                           ensemble_studies=ENSEMBLE_STUDIES, manascore_dir=None,
                           removeNoPositivePatient=False, alternative="greater",
                           sctrp_display_keys=(KEY_CLONO,),# KEY_ONLY, KEY_XGB),
                           show=False):
    """
    Study-level precision@k for one study, all k in a single grouped-bar figure.
    Bar = global top-k precision over the whole study; stars = per-patient paired
    t-test (patients are replicates), so a tall bar can still fail the test.
    scTRP keys resolve per study: ensemble-suffixed for `ensemble_studies`,
    plain otherwise.
    """
    out_dir = fig_dir / study
    out_dir.mkdir(parents=True, exist_ok=True)

    ctx = _prep_precision_ctx(cell_data, study, pool_type, ensemble, ensemble_studies,
                              manascore_dir=manascore_dir,
                              removeNoPositivePatient=removeNoPositivePatient,
                              sctrp_display_keys=sctrp_display_keys, k_vals=k_vals)
    if ctx is None:
        print(f"    [skip] {study}: no data for pool_type={pool_type}")
        return

    mo, pal, lab, ref = ctx["method_order"], ctx["palette"], ctx["labels"], ctx["ref_key"]
    base = f"fig6_precision_k_{study}_{ctx['level']}{ctx['suffix']}"
    _save_ctx_tables(ctx, out_dir, base, alternative)

    stars = defaultdict(dict)
    if ref:
        for dk in mo:
            if dk == ref:
                continue
            for k in k_vals:
                stars[dk][k] = _paired_stats(ctx["per_patient"][(ref, k)],
                                             ctx["per_patient"][(dk, k)],
                                             alternative)["star"]

    n_m, n_k = len(mo), len(k_vals)
    print(f"check here {n_m * n_k * 0.16 + 3}")
    x = np.arange(n_k); bar_w = 0.85 / n_m
    fig, ax = plt.subplots(figsize=(max(7.8, n_m * n_k * 0.16 + 3), 6),
                           constrained_layout=False)

    for i, dk in enumerate(mo):
        offset = (i - n_m / 2 + 0.5) * bar_w
        vals = [ctx["study_prec"][dk][k] for k in k_vals]
        is_sctrp = dk in ctx["sctrp_set"]
        bars = ax.bar(x + offset, vals, width=bar_w * 0.92,
                      color=pal.get(dk, "#aaa"), alpha=1.0 if is_sctrp else 0.72,
                      edgecolor="white", linewidth=0.6,
                      label=lab.get(dk, dk).replace("\n", " "),
                      zorder=3 if is_sctrp else 2)
        for bar, v, k in zip(bars, vals, k_vals):
            if np.isnan(v):
                continue
            if is_sctrp:
                ax.text(bar.get_x() + bar.get_width() / 2, v, f"{v:.0%}",
                        ha="left", va="bottom", rotation=70, rotation_mode="anchor",
                        fontsize=FS["value"]+2,
                        fontweight="bold", color=pal.get(dk, "#333"))
            else:
                s = stars[dk].get(k, "")
                if s and s != "ns":
                    ax.text(bar.get_x() + bar.get_width() / 2, v + 0.006, s,
                            ha="center", va="bottom", fontsize=FS["star"]+2, fontweight="bold",
                            color=pal.get(dk, "#555"))

    if not np.isnan(ctx["baseline"]):
        ax.axhline(ctx["baseline"], color="#888888", lw=1.1, ls="--", zorder=1,
                   label=f"Baseline ({ctx['baseline']:.0%})")

    ax.set_xticks(x)
    ax.set_xticklabels([f"Top-{k}" for k in k_vals], fontsize=FS["tick"]+4)
    ax.tick_params(axis="y", labelsize=FS["tick"]+4)
    ax.set_ylabel("Precision (fraction reactive)", fontsize=FS["axlabel"])
    ax.set_title(f"Precision at K — {study} [{ctx['level']}]{ctx['suffix']}",
                 pad=15, fontweight="bold", fontsize=FS["title"], y=1.10)
    ax.set_ylim(0, ax.get_ylim()[1] * 1.1)
    ax.yaxis.set_major_formatter(mticker.PercentFormatter(xmax=1, decimals=0))
    ax.yaxis.grid(True, lw=0.5, color="#eeeeee", zorder=0)
    for sp in ("top", "right"):
        ax.spines[sp].set_visible(False)

    h, l = ax.get_legend_handles_labels()
    ax.legend(h, l, frameon=False, fontsize=FS["legend"]+2, loc="upper right",
          ncol=3, bbox_to_anchor=(1.05, 1.16),
          columnspacing=0.20, handletextpad=0.4, handlelength=0.8, labelspacing=0.1, borderpad=0.2)
    arrow = ">" if alternative == "greater" else "vs"
    #ax.annotate(f"Significance: paired Wilcoxon on per-patient precision "
    #            f"(n = {ctx['n_patients']}),\nscTRP {arrow} benchmark "
    #            f"({alternative});\n* p<0.05  ** p<0.01  *** p<0.001",
    #            xy=(0.0, -0.22), xycoords="axes fraction",
    #            fontsize=FS["note"]+3, color="gray", va="bottom")

    for ext in ("pdf", "png", "svg"):
        fig.savefig(out_dir / f"{base}.{ext}", dpi=1200)
    if show:
        plt.show()
    plt.close(fig)
    print(f"    saved {base}")




def plot_perpatient_precision_k(cell_data, fig_dir, study, pool_type,
                                k_vals=(10, 15, 20, 25, 30), ensemble=ENSEMBLE,
                                ensemble_studies=ENSEMBLE_STUDIES, manascore_dir=None,
                                removeNoPositivePatient=False, alternative="greater",
                                sctrp_display_keys=(KEY_CLONO,),# KEY_ONLY, KEY_XGB),
                                show=False):
    """
    Per-patient precision@k for one study, all k in a single figure: grouped boxes
    (one box per method within each Top-k group) with per-patient jitter points.
    Plotted values and the paired t-test use the identical per-patient series.
    """
    out_dir = fig_dir / study
    out_dir.mkdir(parents=True, exist_ok=True)

    ctx = _prep_precision_ctx(cell_data, study, pool_type, ensemble, ensemble_studies,
                              manascore_dir=manascore_dir,
                              removeNoPositivePatient=removeNoPositivePatient,
                              sctrp_display_keys=sctrp_display_keys, k_vals=k_vals)
    if ctx is None:
        print(f"    [skip] {study}: no data for pool_type={pool_type}")
        return

    mo, pal, lab, ref = ctx["method_order"], ctx["palette"], ctx["labels"], ctx["ref_key"]
    base = f"fig7_perpatient_precision_k_{study}_{ctx['level']}{ctx['suffix']}"
    _save_ctx_tables(ctx, out_dir, base, alternative)

    if ctx["n_patients"] < 5:
        print(f"    [note] {study}: only {ctx['n_patients']} patients — boxes are "
              f"nearly uninformative; the jitter points carry the signal")

    n_m, n_k = len(mo), len(k_vals)
    x, slot = np.arange(n_k), 0.84 / n_m
    rng = np.random.default_rng(0)
    fig, ax = plt.subplots(figsize=(max(11, n_k * n_m * 0.34 + 2.5), 6.5),
                           constrained_layout=True)

    for i, dk in enumerate(mo):
        offset = (i - n_m / 2 + 0.5) * slot
        pos = x + offset
        data = [ctx["per_patient"][(dk, k)].values.astype(float) for k in k_vals]
        bp = ax.boxplot(data, positions=pos, widths=slot * 0.85, patch_artist=True,
                        showfliers=False, manage_ticks=False,
                        medianprops=dict(color="black", lw=1.2),
                        whiskerprops=dict(lw=0.8), capprops=dict(lw=0.8),
                        boxprops=dict(lw=0.6))
        is_sctrp = dk in ctx["sctrp_set"]
        for patch in bp["boxes"]:
            patch.set_facecolor(pal.get(dk, "#aaa"))
            patch.set_alpha(0.85 if is_sctrp else 0.55)
        for p, vals, k in zip(pos, data, k_vals):
            if not len(vals):
                continue
            ax.scatter(p + rng.uniform(-slot * 0.28, slot * 0.28, size=len(vals)),
                       vals, s=10, color=pal.get(dk, "#555"), alpha=0.85,
                       edgecolor="white", linewidth=0.25, zorder=4)
            if ref and dk != ref:
                s = _paired_stats(ctx["per_patient"][(ref, k)],
                                  ctx["per_patient"][(dk, k)], alternative)["star"]
                if s and s != "ns":
                    ax.text(p, np.nanmax(vals) + 0.025, s, ha="center", va="bottom",
                            fontsize=FS["star"], color=pal.get(dk, "#555"))

    if not np.isnan(ctx["baseline"]):
        ax.axhline(ctx["baseline"], color="#888888", lw=1.1, ls="--", zorder=1)

    handles = [Patch(facecolor=pal.get(dk, "#aaa"),
                     alpha=0.85 if dk in ctx["sctrp_set"] else 0.55,
                     label=lab.get(dk, dk).replace("\n", " ")) for dk in mo]
    if not np.isnan(ctx["baseline"]):
        handles.append(Patch(facecolor="none", edgecolor="#888888",
                             label=f"Baseline ({ctx['baseline']:.0%})"))

    ax.set_xticks(x)
    ax.set_xticklabels([f"Top-{k}" for k in k_vals], fontsize=FS["tick"])
    ax.set_xlim(-0.5, n_k - 0.5)
    ax.tick_params(axis="y", labelsize=FS["tick"])
    ax.set_ylabel("Per-patient precision (fraction reactive)", fontsize=FS["axlabel"])
    ax.set_ylim(-0.03, 1.08)
    ax.yaxis.set_major_formatter(mticker.PercentFormatter(xmax=1, decimals=0))
    ax.yaxis.grid(True, lw=0.5, color="#eeeeee", zorder=0)
    for sp in ("top", "right"):
        ax.spines[sp].set_visible(False)
    ax.set_title(f"Per-patient precision at K — {study}  [{ctx['level']}]{ctx['suffix']}",
                 pad=15, fontweight="bold", fontsize=FS["title"], y=1.15)
    ax.legend(handles=handles, frameon=False, fontsize=FS["legend"],
              loc="upper right", ncol=5, bbox_to_anchor=(1.0, 1.20))
    arrow = ">" if alternative == "greater" else "vs"
    ax.annotate(f"Each point = one patient (n = {ctx['n_patients']}). Significance: "
                f"paired t-test, scTRP {arrow} benchmark ({alternative}); "
                f"* p<0.05  ** p<0.01  *** p<0.001",
                xy=(0.0, -0.12), xycoords="axes fraction",
                fontsize=FS["note"], color="gray", va="bottom")

    for ext in ("pdf", "png", "svg"):
        fig.savefig(out_dir / f"{base}.{ext}", dpi=1200)
    if show:
        plt.show()
    plt.close(fig)
    print(f"    saved {base}")



# ═══════════════════════════════════════════════════════════════════════════
# Per-patient Average Precision (AP) and R-precision (precision@R)
#
# Append this to plot_ablation_study_topKprecision.py (after the precision@k
# functions). It reuses _prep_precision_ctx / _paired_stats / _topn_precision /
# _as_int_reactivity / FS / METHOD_* from that file, so method mapping, clonotype
# dedup, palette, labels and the per-patient paired-Wilcoxon stay identical.
#
# Unlike precision@k, AP and R-precision are ONE scalar per patient (no k axis):
#   AP        : ranking quality over the whole list; self-normalises to each
#               patient's positive count (undefined if 0 or all positives).
#   R-precision: precision within the top-R items, R = that patient's # positives
#               (adapts k per patient; undefined if 0 positives).
# Both are the right answer to heterogeneous per-patient positive counts.
# ═══════════════════════════════════════════════════════════════════════════
from sklearn.metrics import average_precision_score
from matplotlib.lines import Line2D


def _per_patient_ap_series(dfm, score="score", label="reactivity", group="sample"):
    """Per-patient Average Precision. NaN when a patient has 0 or all positives
    (AP is undefined without both classes)."""
    out = {}
    for pat, g in dfm.groupby(group):
        y = _as_int_reactivity(g[label]).to_numpy()
        s = g[score].to_numpy().astype(float)
        m = ~pd.isna(y)
        y, s = y[m].astype(int), s[m]
        npos = int(y.sum())
        out[pat] = float(average_precision_score(y, s)) if 0 < npos < len(y) else np.nan
    return pd.Series(out, dtype=float)


def _per_patient_rprec_series(dfm, score="score", label="reactivity", group="sample"):
    """Per-patient R-precision = precision within the top-R ranked items, where R is
    that patient's own number of positives. NaN when a patient has 0 positives."""
    out = {}
    for pat, g in dfm.groupby(group):
        y = _as_int_reactivity(g[label]).to_numpy()
        s = g[score].to_numpy().astype(float)
        m = ~pd.isna(y)
        y, s = y[m].astype(int), s[m]
        r = int(y.sum())
        out[pat] = _topn_precision(s, y, r) if r > 0 else np.nan
    return pd.Series(out, dtype=float)


def _per_patient_prevalence_series(dfm, label="reactivity", group="sample"):
    """Per-patient positive fraction — the chance level for AP and R-precision."""
    out = {}
    for pat, g in dfm.groupby(group):
        y = _as_int_reactivity(g[label]).dropna().astype(int)
        out[pat] = float(y.mean()) if len(y) else np.nan
    return pd.Series(out, dtype=float)


# metric_name -> (axis label, per-patient series fn, figure-number prefix)
_SCALAR_METRICS = {
    "AP":    ("Average precision (AP)",       _per_patient_ap_series,    "fig8"),
    "Rprec": ("R-precision",     _per_patient_rprec_series, "fig9"),
}


def _save_scalar_tables(ctx, series_by_method, prev_by_method, metric_name,
                        out_dir, base, alternative):
    """Write per-patient values (+ prevalence + lift), a study-level summary, and
    the per-patient paired t-test vs the reference (scTRP)."""
    meta = {"study": ctx["study"], "level": ctx["level"],
            "pool_type": ctx["pool_type"], "metric": metric_name}
    lbl = lambda dk: ctx["labels"].get(dk, dk).replace("\n", " ")

    rows = []
    for dk in ctx["method_order"]:
        s, prev = series_by_method[dk], prev_by_method[dk]
        for pat, v in s.items():
            p = prev.get(pat, np.nan)
            rows.append({**meta, "method": dk, "method_label": lbl(dk), "sample": pat,
                         "value": (float(v) if pd.notna(v) else np.nan),
                         "prevalence": p,
                         "lift": (v / p if (pd.notna(v) and p and p > 0) else np.nan)})
    pd.DataFrame(rows).to_csv(out_dir / f"{base}_per_patient.csv",
                              index=False, float_format="%.6f")

    srows = []
    for dk in ctx["method_order"]:
        s = series_by_method[dk].dropna()
        prev = prev_by_method[dk].reindex(s.index)
        lift = (s / prev).replace([np.inf, -np.inf], np.nan)
        srows.append({**meta, "method": dk, "method_label": lbl(dk),
                      "n_patients_defined": int(s.size),
                      "mean": (float(s.mean()) if s.size else np.nan),
                      "median": (float(s.median()) if s.size else np.nan),
                      "std": (float(s.std(ddof=1)) if s.size > 1 else np.nan),
                      "mean_prevalence": (float(prev.mean()) if prev.size else np.nan),
                      "mean_lift": (float(lift.mean()) if lift.notna().any() else np.nan)})
    pd.DataFrame(srows).to_csv(out_dir / f"{base}_summary.csv",
                               index=False, float_format="%.6f")

    ref = ctx["ref_key"]
    if ref is None:
        return
    ref_s = series_by_method[ref].dropna()
    wrows = []
    for dk in ctx["method_order"]:
        if dk == ref:
            continue
        st = _paired_stats(ref_s, series_by_method[dk].dropna(), alternative)
        wrows.append({**meta, "reference": ref, "method": dk, "method_label": lbl(dk),
                      "test": f"paired t-test ({alternative})", **st})
    pd.DataFrame(wrows).to_csv(out_dir / f"{base}_t_test.csv",
                               index=False, float_format="%.6f")


def _plot_perpatient_scalar(ctx, series_by_method, prev_by_method, metric_name,
                            fig_dir, alternative, show):
    """Horizontal box + per-patient jitter, scTRP at top, benchmarks below.
    Stars = per-patient paired t-test (scTRP vs benchmark); dashed line = chance
    (mean per-patient prevalence)."""
    metric_label, _, fig_no = _SCALAR_METRICS[metric_name]
    out_dir = fig_dir / ctx["study"]
    out_dir.mkdir(parents=True, exist_ok=True)
    base = f"{fig_no}_perpatient_{metric_name}_{ctx['study']}_{ctx['level']}{ctx['suffix']}"
    _save_scalar_tables(ctx, series_by_method, prev_by_method, metric_name,
                        out_dir, base, alternative)

    pal, lab, ref = ctx["palette"], ctx["labels"], ctx["ref_key"]
    sctrp = [m for m in (KEY_CLONO, KEY_ONLY, KEY_XGB) if m in ctx["method_order"]]
    bench = [m for m in ctx["method_order"] if m not in sctrp]
    plot_order = sctrp + bench                       # top -> bottom
    n = len(plot_order)
    positions = np.arange(n)[::-1]                    # first entry at the top

    rng = np.random.default_rng(0)
    fig, ax = plt.subplots(figsize=(7.8, max(4.0, 0.62 * n + 1.7)))

    data = [series_by_method[dk].dropna().values.astype(float) for dk in plot_order]
    bp = ax.boxplot(data, positions=positions, vert=False, widths=0.58,
                    patch_artist=True, showfliers=False, manage_ticks=False,
                    medianprops=dict(color="black", lw=1.6),
                    whiskerprops=dict(lw=0.8), capprops=dict(lw=0.8),
                    boxprops=dict(lw=0.6))
    for patch, dk in zip(bp["boxes"], plot_order):
        patch.set_facecolor(pal.get(dk, "#aaa"))
        patch.set_alpha(0.85 if dk in ctx["sctrp_set"] else 0.55)

    for pos, dk in zip(positions, plot_order):
        vals = series_by_method[dk].dropna().values.astype(float)
        if len(vals):
            ax.scatter(vals, pos + rng.uniform(-0.16, 0.16, size=len(vals)),
                       s=36, color=pal.get(dk, "#555"), alpha=0.9,
                       edgecolor="white", linewidth=0.3, zorder=4)
        if ref and dk != ref:
            st = _paired_stats(series_by_method[ref].dropna(),
                               series_by_method[dk].dropna(), alternative)
            if st["star"] and st["star"] != "ns":
                ax.text(1.04, pos, st["star"], ha="left", va="center",
                        fontsize=FS["star"]+8, color=pal.get(dk, "#555"), fontweight="bold")

    prev_ref = prev_by_method.get(ref) if ref else None
    if prev_ref is None or not prev_ref.dropna().size:
        prev_ref = next(iter(prev_by_method.values()))
    chance = float(prev_ref.dropna().mean()) if prev_ref.dropna().size else np.nan
    if not np.isnan(chance):
        ax.axvline(chance, color="#888888", lw=1.1, ls="--", zorder=1)
        ax.legend(handles=[Line2D([0], [0], color="#888888", lw=1.08, ls="--",
                                  label=f"Baseline ({chance:.0%})")],
                  frameon=False, fontsize=FS["legend"]+6, loc="upper right", bbox_to_anchor=(1.04, 1.04))

    n_def = int(series_by_method[ref].dropna().size) if ref else \
        max((series_by_method[dk].dropna().size for dk in plot_order), default=0)

    ax.set_yticks(positions)
    #ax.set_yticklabels([lab.get(dk, dk).replace("\n", " ") for dk in plot_order],
    ax.set_yticklabels([lab.get(dk, dk) for dk in plot_order],
                       fontsize=FS["tick"]+3)
    ax.set_ylim(-0.7, n - 0.3)
    ax.set_xlim(-0.03, 1.16)
    ax.set_xlabel(metric_label, fontsize=FS["axlabel"])
    ax.xaxis.set_major_formatter(mticker.PercentFormatter(xmax=1, decimals=0))
    ax.tick_params(axis="x", labelsize=FS["tick"]+5)
    ax.xaxis.grid(True, lw=0.5, color="#eeeeee", zorder=0)
    for sp in ("top", "right"):
        ax.spines[sp].set_visible(False)
    ax.set_title(f"{metric_label} per patient - {ctx['study']} ",
                 #f"[{ctx['level']}]{ctx['suffix']}",
                 pad=12, fontweight="bold", fontsize=FS["title"])
    arrow = ">" if alternative == "greater" else "vs"
    
    #ax.annotate(f"Each point = one patient (n = {n_def}).\n"
    #            f"Significance: paired Wilcoxon, scTRP {arrow} benchmark ({alternative});\n"
    #            f"* p<0.05  ** p<0.01  *** p<0.001",
    #            xy=(0.0, -0.23), xycoords="axes fraction",
    #            fontsize=FS["note"]+3, color="gray", va="bottom")

    for ext in ("pdf", "png", "svg"):
        fig.savefig(out_dir / f"{base}.{ext}", dpi=1200, bbox_inches="tight")
    if show:
        plt.show()
    plt.close(fig)
    print(f"    saved {base}")


def _run_scalar_metric(cell_data, fig_dir, study, pool_type, metric_name, *,
                       ensemble=ENSEMBLE, ensemble_studies=ENSEMBLE_STUDIES,
                       manascore_dir=None, removeNoPositivePatient=False,
                       alternative="greater",
                       sctrp_display_keys=(KEY_CLONO,), show=False):
    """Shared driver: build the same ctx as precision@k, compute the per-patient
    scalar for every method, then plot + save."""
    ctx = _prep_precision_ctx(cell_data, study, pool_type, ensemble, ensemble_studies,
                              manascore_dir=manascore_dir,
                              removeNoPositivePatient=removeNoPositivePatient,
                              sctrp_display_keys=sctrp_display_keys, k_vals=())
    if ctx is None:
        print(f"    [skip] {study}: no data for pool_type={pool_type}")
        return
    series_fn = _SCALAR_METRICS[metric_name][1]
    series = {dk: series_fn(ctx["prepped"][dk]) for dk in ctx["method_order"]}
    prev = {dk: _per_patient_prevalence_series(ctx["prepped"][dk])
            for dk in ctx["method_order"]}
    _plot_perpatient_scalar(ctx, series, prev, metric_name, fig_dir, alternative, show)


def plot_perpatient_ap(cell_data, fig_dir, study, pool_type, **kw):
    """Per-patient Average Precision, one figure per study. Same call shape as
    plot_perpatient_precision_k (minus k_vals)."""
    _run_scalar_metric(cell_data, fig_dir, study, pool_type, "AP", **kw)


def plot_perpatient_rprecision(cell_data, fig_dir, study, pool_type, **kw):
    """Per-patient R-precision (precision@R), one figure per study."""
    _run_scalar_metric(cell_data, fig_dir, study, pool_type, "Rprec", **kw)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def plot_single_sample_violin(
    df, sample_id, input_dir, score_col="score", patient_col="sample",
    title=None, figsize=(4, 5), save_prefix=None,
):
    """
    Single violin (no split) for one patient's score distribution.
    Same visual style as plot_split_violin, for cases where reactivity
    labels aren't available (df only has [patient_col, score_col]).
    """
    d = df[df[patient_col] == sample_id][[score_col, patient_col]].dropna().copy()
    if d.empty:
        raise ValueError(f"No rows found for {patient_col} == {sample_id!r}")

    fig, ax = plt.subplots(figsize=figsize)

    sns.violinplot(
        data=d, x=patient_col, y=score_col,
        inner="quartile", density_norm="count", cut=0,
        color="#888888", linewidth=0.8, ax=ax,
    )

    ax.set_xlabel("")
    ax.set_ylabel(score_col)
    ax.set_title(title if title else str(sample_id))
    for lbl in ax.get_xticklabels():
        lbl.set_rotation(45)
        lbl.set_ha("right")
    for s in ("top", "right"):
        ax.spines[s].set_visible(False)
    fig.tight_layout()
    fig.show()

    if save_prefix:
        for ext in ("pdf", "png", "svg"):
            fig.savefig(input_dir / f"{save_prefix}.{ext}", dpi=300, bbox_inches="tight")
    return fig, ax
```

```python
_s2 = sns.color_palette('Set2', n_colors=8)
PAL_BINARY = {'Non_Response': _s2[2], 'Response': _s2[1]}

def plot_split_violin(
    df, score_col="score", patient_col="sample", reactivity_col="reactivity",
    title=None, order=None, figsize=None, save_prefix=None,
    sample_id=None,
):
    SPLIT_PALETTE = {0: PAL_BINARY['Non_Response'], 1: PAL_BINARY['Response']}

    d = df[[score_col, patient_col, reactivity_col]].dropna().copy()
    if sample_id is not None:
        d = d[d[patient_col] == sample_id]
        if d.empty:
            raise ValueError(f"No rows found for {patient_col} == {sample_id!r}")

    d[reactivity_col] = d[reactivity_col].astype(int)

    patients = order if order is not None else sorted(d[patient_col].unique())
    d[patient_col] = pd.Categorical(d[patient_col], categories=patients, ordered=True)
    n_pat = len(patients)

    if figsize is None:
        figsize = (max(3, 0.7 * n_pat + 2), 5)
    fig, ax = plt.subplots(figsize=figsize)

    sns.violinplot(
        data=d, x=patient_col, y=score_col, hue=reactivity_col,
        split=True, inner="quartile", density_norm="count", common_norm=True, cut=0,
        palette=SPLIT_PALETTE, hue_order=[0, 1], order=patients,
        linewidth=0.8, ax=ax,
    )

    ax.set_xlabel("")
    ax.set_ylabel("scTRP prediction score", fontsize=14, fontweight="bold")
    
    if title:
        ax.set_title(title, fontsize=14, fontweight="bold")
    for lbl in ax.get_xticklabels():
        lbl.set_rotation(0)
        lbl.set_ha("center")
    ax.set_xticklabels(["Patient_1"], fontsize=13)
    ax.tick_params(axis="y", labelsize=13)

    handles, _ = ax.get_legend_handles_labels()
    ax.legend(handles, ["Non-reactive", "Reactive"],
              title=None, frameon=False, loc="upper left",
          bbox_to_anchor=(0.55, 1.0), fontsize=12)
    for s in ("top", "right"):
        ax.spines[s].set_visible(False)
    fig.tight_layout()
    fig.show()

    if save_prefix:
        for ext in ("pdf", "png", "svg"):
            fig.savefig(input_dir / f"{save_prefix}.{ext}", dpi=300, bbox_inches="tight")
    return fig, ax
```

```python
from scipy.stats import gaussian_kde
import numpy as np

def check_area_ratio(df, sample_id, score_col="score", patient_col="sample", reactivity_col="reactivity"):
    d = df[df[patient_col] == sample_id]
    g0 = d[d[reactivity_col] == 0][score_col].dropna().values
    g1 = d[d[reactivity_col] == 1][score_col].dropna().values

    n0, n1 = len(g0), len(g1)
    print(f"Raw counts -> reactivity==0: {n0}, reactivity==1: {n1}, ratio: {n0/n1:.2f}")

    # KDE area check (each KDE integrates to 1 by construction, so this
    # just re-confirms n0/n1 -- the real check is whether seaborn's
    # rendered widths integrate proportionally, which we approximate below)
    kde0 = gaussian_kde(g0)
    kde1 = gaussian_kde(g1)
    y_grid = np.linspace(min(g0.min(), g1.min()), max(g0.max(), g1.max()), 2000)
    area0 = np.trapz(kde0(y_grid), y_grid) * n0
    area1 = np.trapz(kde1(y_grid), y_grid) * n1
    print(f"Count-scaled KDE area ratio: {area0/area1:.2f} (target: {n0/n1:.2f})")

check_area_ratio(df, "chiffelle_Patient3_T0")
```
